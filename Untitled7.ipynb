```python
# @title
"""
World Trade Network Visualisation — Python / NetworkX
======================================================
IMPORTER VERSION (TOP-1) — Replicates Appendix 7.3 of De Benedictis et al. (2013),
CEPII Working Paper No. 2013-24, using Python instead of Pajek.

This is the import-side, single-partner counterpart of the original
export-based script: instead of each country's largest EXPORT destination,
Figs 2 & 3 here show each country's SINGLE largest IMPORT source
(its one major supplier).
Figs 6, 7, 8 are unchanged since they are built from full bilateral flows,
not from the top-k selection.

Produces 5 figure types (≈ Figs 2, 3, 6, 7, 8 of the paper)
+ a bonus centrality-heatmap, all saved as high-resolution PNG.

PANEL : 175 countries  |  YEARS : 2005-2024  (one year per run)

REQUIREMENTS (run once):
    pip install networkx matplotlib pandas numpy scipy

HOW TO RUN:
    1. Set BACI_DIR, OUTPUT_DIR, YEAR below.
    2. python wtn_visualisation_importer_top1.py
       — or paste the whole file into a single Colab cell.

STEPS OVERVIEW
--------------
Step 1  Country metadata  – lat/lon, region, colour for each of 175 ISO-3 codes
Step 2  Load BACI         – aggregate product-level flows to bilateral country pairs
Step 3  Build graphs      – (a) top-1 import partner (supplier) binary graph
                            (b) full weighted directed graph
                            (c) threshold sub-graphs (25 / 50 / 75 / 90 %)
                            (d) regional aggregation graph
Step 4  Layouts           – geographic (lon, lat) or Kamada-Kawai spring
Step 5  Draw figures
        Fig 2  Geo layout   : countries at real coordinates, single-supplier arcs
                              (arrow points FROM each country's one major supplier
                               TO the importer — i.e. supplier → importer,
                               matching the real direction of the trade flow)
        Fig 3  KK layout    : force-directed, same single-supplier arcs, structural position
        Fig 6  Thresholds   : 2×2 panel showing trade at 25/50/75/90 % of world total
        Fig 7  Europe zoom  : intra-European links only, KK layout
        Fig 8  Regions      : flows aggregated to 9 macro-regions, circular layout
        Bonus  Heatmap      : centrality measure × country × year (reads saved CSV)

NODE SIZE NOTE
---------------
Because each importer is, by construction, given exactly its single largest
supplier, IN-degree of every active node in Figs 2 & 3 is constant (≈1) and
is not informative. The meaningful "hub" measure here is OUT-degree: a country
with a large out-degree is one that many OTHER countries picked as their
single major supplier — i.e. a major global supplier (think China, USA,
Germany). Figs 2 & 3 below therefore size nodes by OUT-degree, not in-degree.
"""

import os
import glob
import warnings

import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

warnings.filterwarnings("ignore")

# ─────────────────────────────────────────────────────────────────────────────
# CONFIG  ← edit these three lines
# ─────────────────────────────────────────────────────────────────────────────
BACI_DIR   = "/content/baci_data"           # folder with BACI_HS92_Y{year}_V*.csv
OUTPUT_DIR = "/content/figures"             # where to write PNG files
YEAR       = 2024                           # reference year for visualisation

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ─────────────────────────────────────────────────────────────────────────────
# STEP 1 — COUNTRY METADATA
# columns : (cid, numeric_iso, iso3, name, lat, lon, region, hex_colour)
# ─────────────────────────────────────────────────────────────────────────────
COUNTRY_META = [
    # ── Latin America (34) ───────────────────────────────────────────────────
    (  1, 533, "ABW", "Aruba",                12.52,  -69.97, "Latin America",  "#2ca02c"),
    (  7,  32, "ARG", "Argentina",           -38.42,  -63.62, "Latin America",  "#2ca02c"),
    (  9,  28, "ATG", "Antigua and Barbuda",  17.06,  -61.80, "Latin America",  "#2ca02c"),
    ( 22,  84, "BLZ", "Belize",               17.19,  -88.50, "Latin America",  "#2ca02c"),
    ( 24,  68, "BOL", "Bolivia",             -16.29,  -63.59, "Latin America",  "#2ca02c"),
    ( 25,  76, "BRA", "Brazil",              -14.24,  -51.93, "Latin America",  "#2ca02c"),
    ( 26,  52, "BRB", "Barbados",             13.19,  -59.54, "Latin America",  "#2ca02c"),
    ( 32, 152, "CHL", "Chile",               -35.68,  -71.54, "Latin America",  "#2ca02c"),
    ( 37, 170, "COL", "Colombia",              4.57,  -74.30, "Latin America",  "#2ca02c"),
    ( 38, 188, "CRI", "Costa Rica",            9.75,  -83.75, "Latin America",  "#2ca02c"),
    ( 39, 192, "CUB", "Cuba",                 21.52,  -77.78, "Latin America",  "#2ca02c"),
    ( 40, 136, "CYM", "Cayman Islands",       19.31,  -81.25, "Latin America",  "#2ca02c"),
    ( 44, 212, "DMA", "Dominica",             15.41,  -61.37, "Latin America",  "#2ca02c"),
    ( 46, 214, "DOM", "Dominican Republic",   18.74,  -70.16, "Latin America",  "#2ca02c"),
    ( 48, 218, "ECU", "Ecuador",              -1.83,  -78.18, "Latin America",  "#2ca02c"),
    ( 66, 320, "GTM", "Guatemala",            15.78,  -90.23, "Latin America",  "#2ca02c"),
    ( 67, 328, "GUY", "Guyana",                4.86,  -58.93, "Latin America",  "#2ca02c"),
    ( 69, 340, "HND", "Honduras",             15.20,  -86.24, "Latin America",  "#2ca02c"),
    ( 71, 332, "HTI", "Haiti",                18.97,  -72.29, "Latin America",  "#2ca02c"),
    ( 81, 388, "JAM", "Jamaica",              18.11,  -77.30, "Latin America",  "#2ca02c"),
    ( 88, 659, "KNA", "Saint Kitts and Nevis",17.36,  -62.78, "Latin America",  "#2ca02c"),
    ( 95, 662, "LCA", "Saint Lucia",          13.91,  -60.98, "Latin America",  "#2ca02c"),
    (104, 484, "MEX", "Mexico",               23.63, -102.55, "Latin America",  "#2ca02c"),
    (119, 558, "NIC", "Nicaragua",            12.87,  -85.21, "Latin America",  "#2ca02c"),
    (126, 591, "PAN", "Panama",                8.54,  -80.78, "Latin America",  "#2ca02c"),
    (127, 604, "PER", "Peru",                 -9.19,  -75.02, "Latin America",  "#2ca02c"),
    (133, 600, "PRY", "Paraguay",            -23.44,  -58.44, "Latin America",  "#2ca02c"),
    (144, 222, "SLV", "El Salvador",          13.79,  -88.90, "Latin America",  "#2ca02c"),
    (146, 740, "SUR", "Suriname",              3.92,  -56.03, "Latin America",  "#2ca02c"),
    (157, 780, "TTO", "Trinidad and Tobago",  10.69,  -61.22, "Latin America",  "#2ca02c"),
    (164, 858, "URY", "Uruguay",             -32.52,  -55.77, "Latin America",  "#2ca02c"),
    (167, 670, "VCT", "Saint Vincent and the Grenadines", 13.25, -61.20, "Latin America", "#2ca02c"),
    (168, 862, "VEN", "Venezuela",             6.42,  -66.59, "Latin America",  "#2ca02c"),
    (169,  92, "VGB", "Virgin Islands",       18.42,  -64.64, "Latin America",  "#2ca02c"),
    # ── North America (3) ────────────────────────────────────────────────────
    ( 23,  60, "BMU", "Bermuda",              32.32,  -64.75, "North America",  "#f5c518"),
    ( 30, 124, "CAN", "Canada",               56.13, -106.35, "North America",  "#f5c518"),
    (165, 842, "USA", "United States",        37.09,  -95.71, "North America",  "#f5c518"),
    # ── Europe (38) ──────────────────────────────────────────────────────────
    (  4,   8, "ALB", "Albania",              41.15,   20.17, "Europe",         "#d62728"),
    ( 11,  40, "AUT", "Austria",              47.52,   14.55, "Europe",         "#d62728"),
    ( 17, 100, "BGR", "Bulgaria",             42.73,   25.49, "Europe",         "#d62728"),
    ( 19,  70, "BIH", "Bosnia and Herzegovina",43.92,   17.68, "Europe",         "#d62728"),
    ( 20, 112, "BLR", "Belarus",              53.71,   27.95, "Europe",         "#d62728"),
    ( 31, 757, "CHE", "Switzerland",          46.82,    8.23, "Europe",         "#d62728"),
    ( 41, 196, "CYP", "Cyprus",               35.13,   33.43, "Europe",         "#d62728"),
    ( 42, 203, "CZE", "Czech Republic",       49.82,   15.47, "Europe",         "#d62728"),
    ( 43, 276, "DEU", "Germany",              51.17,   10.45, "Europe",         "#d62728"),
    ( 45, 208, "DNK", "Denmark",              56.26,    9.50, "Europe",         "#d62728"),
    ( 50, 724, "ESP", "Spain",                40.46,   -3.75, "Europe",         "#d62728"),
    ( 51, 233, "EST", "Estonia",              58.60,   25.01, "Europe",         "#d62728"),
    ( 53, 246, "FIN", "Finland",              64.00,   26.00, "Europe",         "#d62728"),
    ( 55, 251, "FRA", "France",               46.23,    2.21, "Europe",         "#d62728"),
    ( 58, 826, "GBR", "United Kingdom",       55.38,   -3.44, "Europe",         "#d62728"),
    ( 65, 300, "GRC", "Greece",               39.07,   21.82, "Europe",         "#d62728"),
    ( 70, 191, "HRV", "Croatia",              45.10,   15.20, "Europe",         "#d62728"),
    ( 72, 348, "HUN", "Hungary",              47.16,   19.50, "Europe",         "#d62728"),
    ( 75, 372, "IRL", "Ireland",              53.14,   -8.24, "Europe",         "#d62728"),
    ( 78, 352, "ISL", "Iceland",              64.96,  -19.02, "Europe",         "#d62728"),
    ( 80, 380, "ITA", "Italy",                41.87,   12.57, "Europe",         "#d62728"),
    ( 97, 440, "LTU", "Lithuania",            55.17,   23.88, "Europe",         "#d62728"),
    ( 98, 428, "LVA", "Latvia",               56.88,   24.60, "Europe",         "#d62728"),
    (101, 498, "MDA", "Moldova",              47.41,   28.37, "Europe",         "#d62728"),
    (106, 807, "MKD", "Macedonia",            41.61,   21.75, "Europe",         "#d62728"),
    (108, 470, "MLT", "Malta",                35.94,   14.38, "Europe",         "#d62728"),
    (120, 528, "NLD", "Netherlands",          52.13,    5.29, "Europe",         "#d62728"),
    (121, 579, "NOR", "Norway",               60.47,    8.47, "Europe",         "#d62728"),
    (130, 616, "POL", "Poland",               51.92,   19.15, "Europe",         "#d62728"),
    (132, 620, "PRT", "Portugal",             39.40,   -8.22, "Europe",         "#d62728"),
    (135, 642, "ROM", "Romania",              45.94,   24.97, "Europe",         "#d62728"),
    (136, 643, "RUS", "Russian Federation",   61.52,  105.32, "Europe",         "#d62728"),
    (147, 703, "SVK", "Slovak Republic",      48.67,   19.70, "Europe",         "#d62728"),
    (148, 705, "SVN", "Slovenia",             46.15,   14.99, "Europe",         "#d62728"),
    (149, 752, "SWE", "Sweden",               60.13,   18.64, "Europe",         "#d62728"),
    (163, 804, "UKR", "Ukraine",              48.38,   31.17, "Europe",         "#d62728"),
    (179,  56, "BEL", "Belgium",              50.50,    4.47, "Europe",         "#d62728"),
    (180, 442, "LUX", "Luxembourg",           49.82,    6.13, "Europe",         "#d62728"),
    # ── Middle East (17) ─────────────────────────────────────────────────────
    (  6, 784, "ARE", "United Arab Emirates", 23.42,   53.85, "Middle East",    "#e377c2"),
    (  8,  51, "ARM", "Armenia",              40.07,   45.04, "Middle East",    "#e377c2"),
    ( 12,  31, "AZE", "Azerbaijan",           40.14,   47.58, "Middle East",    "#e377c2"),
    ( 18,  48, "BHR", "Bahrain",              26.07,   50.56, "Middle East",    "#e377c2"),
    ( 59, 268, "GEO", "Georgia",              42.32,   43.36, "Middle East",    "#e377c2"),
    ( 76, 364, "IRN", "Iran",                 32.43,   53.69, "Middle East",    "#e377c2"),
    ( 77, 368, "IRQ", "Iraq",                 33.22,   43.68, "Middle East",    "#e377c2"),
    ( 79, 376, "ISR", "Israel",               31.05,   34.85, "Middle East",    "#e377c2"),
    ( 82, 400, "JOR", "Jordan",               30.59,   36.24, "Middle East",    "#e377c2"),
    ( 90, 414, "KWT", "Kuwait",               29.31,   47.48, "Middle East",    "#e377c2"),
    ( 92, 422, "LBN", "Lebanon",              33.85,   35.86, "Middle East",    "#e377c2"),
    (124, 512, "OMN", "Oman",                 21.51,   55.92, "Middle East",    "#e377c2"),
    (134, 634, "QAT", "Qatar",                25.35,   51.18, "Middle East",    "#e377c2"),
    (138, 682, "SAU", "Saudi Arabia",         23.89,   45.08, "Middle East",    "#e377c2"),
    (151, 760, "SYR", "Syria",                34.80,   38.997,"Middle East",    "#e377c2"),
    (159, 792, "TUR", "Turkey",               38.96,   35.24, "Middle East",    "#e377c2"),
    (173, 887, "YEM", "Yemen",                15.55,   48.52, "Middle East",    "#e377c2"),
    # ── East Asia (19) ───────────────────────────────────────────────────────
    ( 27,  96, "BRN", "Brunei Darussalam",     4.54,  114.73, "East Asia",      "#9467bd"),
    ( 33, 156, "CHN", "China",                35.86,  104.20, "East Asia",      "#9467bd"),
    ( 68, 344, "HKG", "Hong Kong",            22.40,  114.11, "East Asia",      "#9467bd"),
    ( 73, 360, "IDN", "Indonesia",            -0.79,  113.92, "East Asia",      "#9467bd"),
    ( 83, 392, "JPN", "Japan",                36.20,  138.25, "East Asia",      "#9467bd"),
    ( 87, 116, "KHM", "Cambodia",             12.57,  104.99, "East Asia",      "#9467bd"),
    ( 89, 410, "KOR", "South Korea",          35.91,  127.77, "East Asia",      "#9467bd"),
    ( 91, 418, "LAO", "Laos",                 19.86,  102.50, "East Asia",      "#9467bd"),
    ( 99, 446, "MAC", "Macau",                22.20,  113.54, "East Asia",      "#9467bd"),
    (109, 104, "MMR", "Myanmar",              21.91,   95.96, "East Asia",      "#9467bd"),
    (110, 496, "MNG", "Mongolia",             46.86,  103.85, "East Asia",      "#9467bd"),
    (115, 458, "MYS", "Malaysia",              4.21,  108.04, "East Asia",      "#9467bd"),
    (128, 608, "PHL", "Philippines",          12.88,  121.77, "East Asia",      "#9467bd"),
    (129, 598, "PNG", "Papua New Guinea",     -6.31,  143.96, "East Asia",      "#9467bd"),
    (131, 408, "PRK", "North Korea",          40.34,  127.51, "East Asia",      "#9467bd"),
    (141, 702, "SGP", "Singapore",             1.35,  103.82, "East Asia",      "#9467bd"),
    (154, 764, "THA", "Thailand",             15.87,  100.99, "East Asia",      "#9467bd"),
    (160, 490, "TWN", "Taiwan",               23.70,  120.96, "East Asia",      "#9467bd"),
    (170, 704, "VNM", "Viet Nam",             14.06,  108.28, "East Asia",      "#9467bd"),
    # ── South Asia (8) ───────────────────────────────────────────────────────
    (  2,   4, "AFG", "Afghanistan",          33.94,   67.71, "South Asia",     "#ff7f0e"),
    ( 16,  50, "BGD", "Bangladesh",           23.68,   90.36, "South Asia",     "#ff7f0e"),
    ( 28,  64, "BTN", "Bhutan",               27.51,   90.43, "South Asia",     "#ff7f0e"),
    ( 74, 699, "IND", "India",                20.59,   78.96, "South Asia",     "#ff7f0e"),
    ( 96, 144, "LKA", "Sri Lanka",             7.87,   80.77, "South Asia",     "#ff7f0e"),
    (103, 462, "MDV", "Maldives",              3.20,   73.22, "South Asia",     "#ff7f0e"),
    (122, 524, "NPL", "Nepal",                28.39,   84.12, "South Asia",     "#ff7f0e"),
    (125, 586, "PAK", "Pakistan",              30.38,   69.35, "South Asia",     "#ff7f0e"),
    # ── Oceania (9) ──────────────────────────────────────────────────────────
    ( 10,  36, "AUS", "Australia",           -25.27,  133.78, "Oceania",        "#1f77b4"),
    ( 54, 242, "FJI", "Fiji",                -17.71,  178.07, "Oceania",        "#1f77b4"),
    ( 56, 583, "FSM", "Micronesia",            7.43,  150.55, "Oceania",        "#1f77b4"),
    (105, 584, "MHL", "Marshall Islands",      7.13,  171.18, "Oceania",        "#1f77b4"),
    (116, 540, "NCL", "New Caledonia",       -20.90,  165.62, "Oceania",        "#1f77b4"),
    (123, 554, "NZL", "New Zealand",         -40.90,  174.89, "Oceania",        "#1f77b4"),
    (142,  90, "SLB", "Solomon Islands",      -9.65,  160.16, "Oceania",        "#1f77b4"),
    (171, 548, "VUT", "Vanuatu",             -15.38,  166.96, "Oceania",        "#1f77b4"),
    (172, 882, "WSM", "Samoa",               -13.76, -172.10, "Oceania",        "#1f77b4"),
    # ── Central Asia (5) ─────────────────────────────────────────────────────
    ( 84, 398, "KAZ", "Kazakhstan",           48.02,   66.92, "Central Asia",   "#8c564b"),
    ( 86, 417, "KGZ", "Kyrgyzstan",           41.20,   74.77, "Central Asia",   "#8c564b"),
    (155, 762, "TJK", "Tajikistan",           38.86,   71.28, "Central Asia",   "#8c564b"),
    (156, 795, "TKM", "Turkmenistan",         38.97,   59.56, "Central Asia",   "#8c564b"),
    (166, 860, "UZB", "Uzbekistan",           41.38,   64.59, "Central Asia",   "#8c564b"),
    # ── Africa (42) ──────────────────────────────────────────────────────────
    (  3,  24, "AGO", "Angola",              -11.20,   17.87, "Africa",         "#17becf"),
    ( 13, 108, "BDI", "Burundi",              -3.37,   29.92, "Africa",         "#17becf"),
    ( 14, 204, "BEN", "Benin",                 9.31,    2.32, "Africa",         "#17becf"),
    ( 15, 854, "BFA", "Burkina Faso",         12.24,   -1.56, "Africa",         "#17becf"),
    ( 29, 140, "CAF", "Central African Republic", 6.61, 20.94,"Africa",         "#17becf"),
    ( 34, 384, "CIV", "Cote d'Ivoire",         7.54,   -5.55, "Africa",         "#17becf"),
    ( 35, 120, "CMR", "Cameroon",              7.37,   12.35, "Africa",         "#17becf"),
    ( 36, 178, "COG", "Congo",                -0.23,   15.83, "Africa",         "#17becf"),
    ( 47,  12, "DZA", "Algeria",              28.03,    1.66, "Africa",         "#17becf"),
    ( 49, 818, "EGY", "Egypt",                26.82,   30.80, "Africa",         "#17becf"),
    ( 52, 231, "ETH", "Ethiopia",              9.15,   40.49, "Africa",         "#17becf"),
    ( 57, 266, "GAB", "Gabon",                -0.80,   11.61, "Africa",         "#17becf"),
    ( 60, 288, "GHA", "Ghana",                 7.95,   -1.02, "Africa",         "#17becf"),
    ( 61, 324, "GIN", "Guinea",                9.95,   -9.70, "Africa",         "#17becf"),
    ( 62, 270, "GMB", "Gambia",               13.44,  -15.31, "Africa",         "#17becf"),
    ( 63, 624, "GNB", "Guinea-Bissau",        11.80,  -15.18, "Africa",         "#17becf"),
    ( 64, 226, "GNQ", "Equatorial Guinea",     1.65,   10.27, "Africa",         "#17becf"),
    ( 85, 404, "KEN", "Kenya",                -0.02,   37.91, "Africa",         "#17becf"),
    ( 93, 430, "LBR", "Liberia",               6.43,   -9.43, "Africa",         "#17becf"),
    ( 94, 434, "LBY", "Libya",                26.34,   17.23, "Africa",         "#17becf"),
    (100, 504, "MAR", "Morocco",              31.79,   -7.09, "Africa",         "#17becf"),
    (102, 450, "MDG", "Madagascar",          -18.77,   46.87, "Africa",         "#17becf"),
    (107, 466, "MLI", "Mali",                 17.57,   -3.99, "Africa",         "#17becf"),
    (111, 508, "MOZ", "Mozambique",          -18.67,   35.53, "Africa",         "#17becf"),
    (112, 478, "MRT", "Mauritania",           21.01,  -10.94, "Africa",         "#17becf"),
    (113, 480, "MUS", "Mauritius",           -20.35,   57.55, "Africa",         "#17becf"),
    (114, 454, "MWI", "Malawi",              -13.25,   34.30, "Africa",         "#17becf"),
    (117, 562, "NER", "Niger",                17.61,    8.08, "Africa",         "#17becf"),
    (118, 566, "NGA", "Nigeria",               9.08,    8.68, "Africa",         "#17becf"),
    (137, 646, "RWA", "Rwanda",               -1.94,   29.87, "Africa",         "#17becf"),
    (140, 686, "SEN", "Senegal",              14.50,  -14.45, "Africa",         "#17becf"),
    (143, 694, "SLE", "Sierra Leone",          8.46,  -11.78, "Africa",         "#17becf"),
    (145, 706, "SOM", "Somalia",               5.15,   46.20, "Africa",         "#17becf"),
    (150, 690, "SYC", "Seychelles",           -4.68,   55.49, "Africa",         "#17becf"),
    (152, 148, "TCD", "Chad",                 15.45,   18.73, "Africa",         "#17becf"),
    (153, 768, "TGO", "Togo",                  8.62,    0.82, "Africa",         "#17becf"),
    (158, 788, "TUN", "Tunisia",              33.89,    9.54, "Africa",         "#17becf"),
    (161, 834, "TZA", "Tanzania",             -6.37,   34.89, "Africa",         "#17becf"),
    (162, 800, "UGA", "Uganda",                1.37,   32.29, "Africa",         "#17becf"),
    (176, 180, "ZAR", "Congo Dem. Rep.",      -4.04,   21.76, "Africa",         "#17becf"),
    (177, 894, "ZMB", "Zambia",              -13.13,   27.85, "Africa",         "#17becf"),
    (178, 716, "ZWE", "Zimbabwe",            -19.02,   29.15, "Africa",         "#17becf"),
]

# Convenience lookups
ISO3_META   = {row[2]: row for row in COUNTRY_META}          # iso3 → full row
NUM_TO_ISO  = {row[1]: row[2] for row in COUNTRY_META}       # numeric → iso3
PANEL_ISO3  = [row[2] for row in COUNTRY_META]

REGION_COLORS = {
    "Europe":        "#d62728",
    "North America": "#f5c518",
    "Latin America": "#2ca02c",
    "East Asia":     "#9467bd",
    "South Asia":    "#ff7f0e",
    "Oceania":       "#1f77b4",
    "Central Asia":  "#8c564b",
    "Africa":        "#17becf",
    "Middle East":   "#e377c2",
}


# ─────────────────────────────────────────────────────────────────────────────
# STEP 2 — LOAD BACI  (identical logic to the centrality script)
# ─────────────────────────────────────────────────────────────────────────────
def load_bilateral(year: int) -> pd.DataFrame:
    """
    Read BACI CSV for `year`, filter to the 175-country panel,
    aggregate product-level flows to bilateral totals.
    Returns DataFrame with columns [exp, imp, val].
    """
    pattern = os.path.join(BACI_DIR, f"BACI_HS92_Y{year}_V*.csv")
    files   = glob.glob(pattern)
    if not files:
        raise FileNotFoundError(f"No BACI file for {year} — pattern: {pattern}")

    df = pd.read_csv(files[0], dtype={"k": str, "i": int, "j": int, "v": float})
    df["exp"] = df["i"].map(lambda x: NUM_TO_ISO.get(x))
    df["imp"] = df["j"].map(lambda x: NUM_TO_ISO.get(x))
    df = df.dropna(subset=["exp", "imp"])
    df = df[df["exp"] != df["imp"]]

    bil = (df.groupby(["exp", "imp"])["v"]
             .sum().reset_index()
             .rename(columns={"v": "val"}))
    print(f"  {len(bil):,} bilateral flows | "
          f"{bil['exp'].nunique()} exporters | {bil['imp'].nunique()} importers")
    return bil


# ─────────────────────────────────────────────────────────────────────────────
# STEP 3 — BUILD GRAPHS
# ─────────────────────────────────────────────────────────────────────────────
def build_full_graph(bilateral: pd.DataFrame) -> nx.DiGraph:
    """Full weighted directed graph (all bilateral flows, 175 nodes)."""
    G = nx.DiGraph()
    G.add_nodes_from(PANEL_ISO3)
    for _, r in bilateral.iterrows():
        if r["exp"] in ISO3_META and r["imp"] in ISO3_META:
            G.add_edge(r["exp"], r["imp"], weight=float(r["val"]))
    return G


def build_top_k_importer_graph(bilateral: pd.DataFrame, k: int = 1) -> nx.DiGraph:
    """
    Binary directed graph keeping only each country's top-k IMPORT
    sources (its k largest suppliers). Default k=1 → each importer keeps
    only its single largest supplier.

    For every importer j, we group the bilateral flows by "imp" and keep
    the k rows with the largest trade value — i.e. j's k biggest suppliers.
    An edge is then added supplier → importer (exp → imp), exactly matching
    the real direction of the trade flow.

    This mirrors the "major two importing partners" compression scheme
    used for the sectoral networks (bananas, cement, movies, oil, footwear,
    engines — Figs 10-15) in the paper, applied here to the full aggregate
    World Trade Network instead of a single product.
    """
    G = nx.DiGraph()
    G.add_nodes_from(PANEL_ISO3)
    for imp, grp in bilateral.groupby("imp"):
        for _, r in grp.nlargest(k, "val").iterrows():
            G.add_edge(r["exp"], r["imp"], weight=1.0)
    return G


def build_threshold_graph(bilateral: pd.DataFrame,
                           threshold: float) -> nx.DiGraph:
    """
    Keep the strongest links that cumulatively account for
    `threshold` share of total world trade.
    """
    bil_s = bilateral.sort_values("val", ascending=False).copy()
    bil_s["cumshare"] = bil_s["val"].cumsum() / bil_s["val"].sum()
    subset = bil_s[bil_s["cumshare"].shift(1, fill_value=0.0) < threshold]

    G = nx.DiGraph()
    G.add_nodes_from(PANEL_ISO3)
    for _, r in subset.iterrows():
        if r["exp"] in ISO3_META and r["imp"] in ISO3_META:
            G.add_edge(r["exp"], r["imp"], weight=float(r["val"]))
    return G


# ─────────────────────────────────────────────────────────────────────────────
# STEP 4 — NODE-ATTRIBUTE HELPERS
# ─────────────────────────────────────────────────────────────────────────────
def geo_positions() -> dict:
    """Geographic layout: {iso3: (longitude, latitude)}."""
    return {row[2]: (row[5], row[4]) for row in COUNTRY_META}


def node_colours(nodes) -> list:
    return [ISO3_META[n][7] for n in nodes]


def size_by_indegree(G, nodes, base=50, scale=600) -> list:
    max_d = max(G.in_degree(n) for n in nodes) or 1
    return [base + scale * G.in_degree(n) / max_d for n in nodes]


def size_by_outdegree(G, nodes, base=50, scale=600) -> list:
    max_d = max(G.out_degree(n) for n in nodes) or 1
    return [base + scale * G.out_degree(n) / max_d for n in nodes]


def _legend_handles():
    return [mpatches.Patch(color=c, label=r)
            for r, c in REGION_COLORS.items()]


def _deoverlap_positions(pos: dict, min_dist: float = 0.06, max_iter: int = 300) -> dict:
    """
    Nudge apart any two nodes whose layout coordinates are closer than
    `min_dist`. Kamada-Kawai (and other force-directed layouts) have no
    built-in repulsion for nodes that are weakly or never connected to the
    rest of the graph — isolated/near-isolated nodes get strung along an
    arc by the solver, and that arc can land almost exactly on top of an
    unrelated hub node purely by coincidence. This is a label/marker-only
    fix: it does not touch the graph structure, only the drawn position.
    """
    nodes  = list(pos.keys())
    coords = np.array([pos[n] for n in nodes], dtype=float)
    rng    = np.random.default_rng(0)

    for _ in range(max_iter):
        moved = False
        for i in range(len(nodes)):
            for j in range(i + 1, len(nodes)):
                diff = coords[i] - coords[j]
                d    = np.linalg.norm(diff)
                if d < min_dist:
                    direction = diff if d > 1e-9 else rng.normal(size=2)
                    direction = direction / (np.linalg.norm(direction) + 1e-9)
                    shift = (min_dist - d) / 2
                    coords[i] += direction * shift
                    coords[j] -= direction * shift
                    moved = True
        if not moved:
            break
    return {n: tuple(coords[i]) for i, n in enumerate(nodes)}


def _save(fig, fname: str):
    path = os.path.join(OUTPUT_DIR, fname)
    fig.savefig(path, dpi=180, bbox_inches="tight")
    print(f"  ✓  Saved → {path}")


# ─────────────────────────────────────────────────────────────────────────────
# STEP 5a — FIGURE 2  Geographical layout  (2 MAJOR IMPORTERS / suppliers)
# ─────────────────────────────────────────────────────────────────────────────
def fig2_geographical(bilateral: pd.DataFrame, year: int):
    """
    Countries are pinned to their real (lon, lat) coordinates.
    Arcs show each country's single largest import source (its main
    supplier); arrows point supplier → importer, the real direction of
    the trade flow.
    Node size ∝ OUT-degree (how many other countries picked this country
    as their single major supplier — i.e. its importance as a global
    supplier hub). In-degree is not used here because every active importer
    has, by construction, in-degree ≈ 1.

    Paper analogue: Figure 2 (export side) → import-side counterpart
    """
    G    = build_top_k_importer_graph(bilateral, k=1)
    pos  = geo_positions()
    nds  = list(G.nodes())
    szs  = size_by_outdegree(G, nds)
    cols = node_colours(nds)

    fig, ax = plt.subplots(figsize=(16, 9))
    ax.set_facecolor("#eef2f7")
    fig.patch.set_facecolor("#eef2f7")

    # ── edges ──
    nx.draw_networkx_edges(
        G, pos, ax=ax,
        edge_color="#777777", alpha=0.35,
        arrows=True, arrowsize=8, width=0.7,
        connectionstyle="arc3,rad=0.12",
    )
    # ── nodes ──
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=szs,
                           node_color=cols, alpha=0.88, linewidths=0.4,
                           edgecolors="white")
    # ── labels ──
    nx.draw_networkx_labels(G, pos, ax=ax, font_size=5.5,
                            font_weight="bold", font_color="#111111")

    ax.legend(handles=_legend_handles(), loc="lower left",
              fontsize=7, title="Region", title_fontsize=8,
              framealpha=0.85)
    ax.set_title(
        f"World Trade Network — Geographical Layout\n"
        f"(Major import partner / supplier per country, {year}; "
        f"node size ∝ out-degree)",
        fontsize=12, fontweight="bold",
    )
    ax.set_xlabel("Longitude", fontsize=9)
    ax.set_ylabel("Latitude",  fontsize=9)
    plt.tight_layout()
    _save(fig, f"fig2_geo_import_top1_{year}.png")
    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# STEP 5b — FIGURE 3  Force-directed (Kamada-Kawai) layout  (2 MAJOR IMPORTERS)
# ─────────────────────────────────────────────────────────────────────────────
def fig3_force_directed(bilateral: pd.DataFrame, year: int):
    """
    Same single-supplier binary network as Fig 2 but the spring-
    minimisation (Kamada-Kawai) places structurally central countries
    (the major global suppliers) at the core.

    Paper analogue: Figure 3 (export side) → import-side counterpart
    """
    G    = build_top_k_importer_graph(bilateral, k=1)
    Gu   = G.to_undirected()
    pos  = nx.kamada_kawai_layout(Gu, scale=2.5)
    pos  = _deoverlap_positions(pos)
    nds  = list(G.nodes())
    szs  = size_by_outdegree(G, nds)
    cols = node_colours(nds)

    fig, ax = plt.subplots(figsize=(14, 11))
    ax.set_facecolor("#fafafa")

    nx.draw_networkx_edges(
        G, pos, ax=ax,
        edge_color="#999999", alpha=0.40,
        arrows=True, arrowsize=10, width=0.9,
        connectionstyle="arc3,rad=0.08",
    )
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=szs,
                           node_color=cols, alpha=0.90,
                           linewidths=0.5, edgecolors="white")
    nx.draw_networkx_labels(G, pos, ax=ax, font_size=6.5,
                            font_weight="bold", font_color="#111111")

    ax.legend(handles=_legend_handles(), loc="lower left",
              fontsize=7, title="Region", title_fontsize=8, framealpha=0.85)
    ax.set_title(
        f"World Trade Network — Kamada-Kawai Force Layout\n"
        f"(Major import partner / supplier, {year}; "
        f"major global suppliers cluster at core)",
        fontsize=12, fontweight="bold",
    )
    ax.axis("off")
    plt.tight_layout()
    _save(fig, f"fig3_kk_import_top1_{year}.png")
    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# STEP 5c — FIGURE 6  Trade thresholds (25 / 50 / 75 / 90 %)
# ─────────────────────────────────────────────────────────────────────────────
def fig6_trade_thresholds(bilateral: pd.DataFrame, year: int):
    """
    2×2 panel.  Each sub-figure adds more bilateral links until the
    displayed links account for 25, 50, 75 or 90 % of world trade.
    All panels share the same Kamada-Kawai positions (from the full graph).
    Built from full bilateral flows — unaffected by the export/import
    choice used in Figs 2 & 3.

    Paper analogue: Figure 6
    """
    # Shared layout from the full graph
    G_full = build_full_graph(bilateral)
    pos    = nx.kamada_kawai_layout(G_full.to_undirected(), scale=2.5)
    pos    = _deoverlap_positions(pos)

    thresholds   = [0.25, 0.50, 0.75, 0.90]
    panel_titles = ["(a) 25% of world trade", "(b) 50% of world trade",
                    "(c) 75% of world trade", "(d) 90% of world trade"]

    fig, axes = plt.subplots(2, 2, figsize=(18, 14))
    fig.patch.set_facecolor("#f5f5f5")

    bil_sorted = bilateral.sort_values("val", ascending=False).copy()
    total      = bil_sorted["val"].sum()
    bil_sorted["cumshare"] = bil_sorted["val"].cumsum() / total

    for ax, thr, title in zip(axes.flatten(), thresholds, panel_titles):
        # select flows up to this threshold
        subset = bil_sorted[bil_sorted["cumshare"].shift(1, fill_value=0.0) < thr]

        G_sub = nx.DiGraph()
        G_sub.add_nodes_from(PANEL_ISO3)
        for _, r in subset.iterrows():
            if r["exp"] in ISO3_META and r["imp"] in ISO3_META:
                G_sub.add_edge(r["exp"], r["imp"], weight=float(r["val"]))

        nds  = PANEL_ISO3
        szs  = size_by_indegree(G_sub, nds, base=20, scale=350)
        cols = node_colours(nds)

        # edge grey-scale proportional to log-weight
        edges = list(G_sub.edges(data=True))
        if edges:
            lw_max = np.log1p(max(d["weight"] for _, _, d in edges))
            widths = [0.3 + 2.5 * np.log1p(d["weight"]) / lw_max
                      for _, _, d in edges]
        else:
            widths = []

        ax.set_facecolor("#fafafa")
        if edges:
            nx.draw_networkx_edges(G_sub, pos, ax=ax,
                                   edge_color="#888888",
                                   width=widths, alpha=0.35,
                                   arrows=False)
        nx.draw_networkx_nodes(G_sub, pos, ax=ax, node_size=szs,
                               node_color=cols, alpha=0.88,
                               linewidths=0.4, edgecolors="white")
        nx.draw_networkx_labels(G_sub, pos, ax=ax, font_size=4.5,
                                font_weight="bold")

        active = G_sub.number_of_edges()
        ax.set_title(f"{title}\n({active} links | "
                     f"{sum(1 for n in PANEL_ISO3 if G_sub.degree(n)>0)} countries)",
                     fontsize=10, fontweight="bold")
        ax.axis("off")

    fig.legend(handles=_legend_handles(), loc="lower center",
               ncol=4, fontsize=8, title="Geographical Region",
               title_fontsize=9, framealpha=0.9)
    fig.suptitle(
        f"World Trade Network at Different Cumulative Trade Thresholds, {year}\n"
        f"(Node size ∝ in-degree; edge width ∝ log trade value)",
        fontsize=13, fontweight="bold",
    )
    plt.tight_layout(rect=[0, 0.06, 1, 0.97])
    _save(fig, f"fig6_thresholds_{year}.png")
    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# STEP 5d — FIGURE 7  Zoom into European countries
# ─────────────────────────────────────────────────────────────────────────────
def fig7_europe_zoom(bilateral: pd.DataFrame, year: int):
    """
    Extract the intra-European sub-network and apply its own KK layout.
    Edge width encodes log trade volume. Built from full bilateral flows —
    unaffected by the export/import choice used in Figs 2 & 3.

    Paper analogue: Figure 7
    """
    eu = [row[2] for row in COUNTRY_META if row[6] == "Europe"]
    bil_eu = bilateral[
        bilateral["exp"].isin(eu) & bilateral["imp"].isin(eu)
    ]

    G_eu = nx.DiGraph()
    G_eu.add_nodes_from(eu)
    for _, r in bil_eu.iterrows():
        G_eu.add_edge(r["exp"], r["imp"], weight=float(r["val"]))

    pos  = nx.kamada_kawai_layout(G_eu.to_undirected(), scale=2.5)
    pos  = _deoverlap_positions(pos)
    nds  = eu
    szs  = size_by_indegree(G_eu, nds, base=60, scale=700)

    edges = list(G_eu.edges(data=True))
    if edges:
        lw_max = np.log1p(max(d["weight"] for _, _, d in edges))
        widths = [0.4 + 4.0 * np.log1p(d["weight"]) / lw_max
                  for _, _, d in edges]
        alphas = [0.20 + 0.55 * np.log1p(d["weight"]) / lw_max
                  for _, _, d in edges]
    else:
        widths, alphas = [], []

    fig, ax = plt.subplots(figsize=(14, 11))
    ax.set_facecolor("#fafafa")

    for (u, v, d), w, a in zip(edges, widths, alphas):
        nx.draw_networkx_edges(G_eu, pos, ax=ax,
                               edgelist=[(u, v)],
                               edge_color="#555555",
                               width=w, alpha=a,
                               arrows=True, arrowsize=10,
                               connectionstyle="arc3,rad=0.06")

    nx.draw_networkx_nodes(G_eu, pos, ax=ax, node_size=szs,
                           node_color="#d62728", alpha=0.88,
                           linewidths=0.5, edgecolors="white")
    nx.draw_networkx_labels(G_eu, pos, ax=ax, font_size=7.5,
                            font_weight="bold")

    ax.set_title(
        f"Trade Links Within Europe, {year}\n"
        f"(Node size ∝ in-degree; edge width ∝ log trade value; "
        f"KK layout)",
        fontsize=12, fontweight="bold",
    )
    ax.axis("off")
    plt.tight_layout()
    _save(fig, f"fig7_europe_{year}.png")
    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# STEP 5e — FIGURE 8  Aggregated by geographical regions
# ─────────────────────────────────────────────────────────────────────────────
def fig8_regional(bilateral: pd.DataFrame, year: int):
    """
    Collapse countries to 9 macro-regions.  Each arc's weight is the
    sum of all bilateral flows between the two regions.
    Self-loops represent intra-regional trade. Built from full bilateral
    flows — unaffected by the export/import choice used in Figs 2 & 3.

    Paper analogue: Figure 8
    """
    iso_reg = {row[2]: row[6] for row in COUNTRY_META}
    bil = bilateral.copy()
    bil["exp_reg"] = bil["exp"].map(iso_reg)
    bil["imp_reg"] = bil["imp"].map(iso_reg)
    bil = bil.dropna(subset=["exp_reg", "imp_reg"])

    reg_trade = bil.groupby(["exp_reg", "imp_reg"])["val"].sum().reset_index()

    regions = list(REGION_COLORS.keys())
    G_reg   = nx.DiGraph()
    G_reg.add_nodes_from(regions)
    for _, r in reg_trade.iterrows():
        if r["exp_reg"] in G_reg and r["imp_reg"] in G_reg:
            if G_reg.has_edge(r["exp_reg"], r["imp_reg"]):
                G_reg[r["exp_reg"]][r["imp_reg"]]["weight"] += r["val"]
            else:
                G_reg.add_edge(r["exp_reg"], r["imp_reg"],
                               weight=float(r["val"]))

    # Circular layout for regions
    pos = nx.circular_layout(G_reg, scale=2.0)

    wmax   = max(d["weight"] for _, _, d in G_reg.edges(data=True))
    node_sz = [200 + 2800 * G_reg.in_degree(n, weight="weight") / wmax
               for n in regions]
    colours = [REGION_COLORS[n] for n in regions]

    edges  = [(u, v, d) for u, v, d in G_reg.edges(data=True) if u != v]
    widths = [0.5 + 7.0 * d["weight"] / wmax for _, _, d in edges]

    fig, ax = plt.subplots(figsize=(13, 10))
    ax.set_facecolor("#fafafa")

    nx.draw_networkx_edges(G_reg, pos, ax=ax,
                           edgelist=[(u, v) for u, v, _ in edges],
                           edge_color="#666666",
                           width=widths, alpha=0.50,
                           arrows=True, arrowsize=18,
                           connectionstyle="arc3,rad=0.22")
    nx.draw_networkx_nodes(G_reg, pos, ax=ax, node_size=node_sz,
                           node_color=colours, alpha=0.90,
                           linewidths=0.8, edgecolors="white")
    nx.draw_networkx_labels(G_reg, pos, ax=ax, font_size=8.5,
                            font_weight="bold")

    ax.set_title(
        f"Trade Among Geographical Regions, {year}\n"
        f"(Node size ∝ weighted in-strength; "
        f"edge width ∝ inter-regional flow; self-loops omitted)",
        fontsize=12, fontweight="bold",
    )
    ax.axis("off")
    plt.tight_layout()
    _save(fig, f"fig8_regional_{year}.png")
    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# BONUS — Centrality heatmap over time  (reads saved centrality CSV)
# ─────────────────────────────────────────────────────────────────────────────
def plot_centrality_heatmap(
    csv_path: str,
    measure: str = "In-eigenvector",
    top_n:   int = 20,
):
    """
    Heat-map: countries (rows) × years (columns), colour = centrality value.
    Shows the top-N countries by time-average of the chosen measure.
    Requires the CSV produced by the companion centrality script.

    For the importer-side story, In-eigenvector / In-strength / In-degree
    measures are usually the most relevant (they capture how central a
    country is as a destination of others' major export flows — i.e. how
    much of a magnet it is for suppliers).

    Usage:
        plot_centrality_heatmap(
            "/content/drive/MyDrive/centrality_measures_2005_2024.csv",
            measure="In-eigenvector", top_n=20,
        )
    """
    df  = pd.read_csv(csv_path, sep=";")
    piv = df.pivot(index="iso3", columns="t", values=measure).fillna(0.0)
    top = piv.mean(axis=1).nlargest(top_n).index
    piv = piv.loc[top]

    fig, ax = plt.subplots(figsize=(max(10, len(piv.columns) * 0.55), 6))
    im = ax.imshow(piv.values, aspect="auto", cmap="YlOrRd")
    ax.set_xticks(range(len(piv.columns)))
    ax.set_xticklabels(piv.columns, rotation=45, ha="right", fontsize=8)
    ax.set_yticks(range(len(top)))
    ax.set_yticklabels(top, fontsize=9)
    plt.colorbar(im, ax=ax, label=measure, shrink=0.8)
    ax.set_title(f"Top-{top_n} Countries · {measure} · 2005–2024",
                 fontsize=12, fontweight="bold")
    plt.tight_layout()
    _save(fig, f"heatmap_{measure.replace(' ', '-')}.png")
    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# MAIN
# ─────────────────────────────────────────────────────────────────────────────
if __name__ == "__main__":
    print("=" * 62)
    print(f"  WTN Visualisation (IMPORTER VERSION — TOP-1)  |  {YEAR}  |  175-country panel")
    print("=" * 62)

    print(f"\n  Loading BACI {YEAR} …")
    bilateral = load_bilateral(YEAR)

    print("\n  [1/5] Fig 2 — Geographical layout (single major import partner) …")
    fig2_geographical(bilateral, YEAR)

    print("\n  [2/5] Fig 3 — Kamada-Kawai force layout (single major import partner) …")
    fig3_force_directed(bilateral, YEAR)

    print("\n  [3/5] Fig 6 — Trade-threshold panels (25/50/75/90 %) …")
    fig6_trade_thresholds(bilateral, YEAR)

    print("\n  [4/5] Fig 7 — Europe zoom …")
    fig7_europe_zoom(bilateral, YEAR)

    print("\n  [5/5] Fig 8 — Regional aggregation …")
    fig8_regional(bilateral, YEAR)

    # ── Optional heatmap (uncomment after running the centrality script) ──
    # CSV = "/content/drive/MyDrive/centrality_measures_2005_2024.csv"
    # plot_centrality_heatmap(CSV, measure="In-eigenvector", top_n=20)
    # plot_centrality_heatmap(CSV, measure="W-In-closenness", top_n=20)

    print("\n" + "=" * 62)
    print(f"  All figures → {OUTPUT_DIR}")
    print("=" * 62)
```
