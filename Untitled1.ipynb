```python
"""
World Trade Network Centrality Measures — CORRECTED VERSION
============================================================
Based on De Benedictis et al. (2013), CEPII Working Paper No. 2013-24
Paste this entire file into ONE Google Colab cell and run.

4 BUGS FIXED vs the previous version:
  BUG 1 → Out-strength / In-strength were NOT divided by (N-1=177)
  BUG 2 → Weighted closeness omega used N (countries) instead of M (edges)
           Stata appendix code: omega = w / mean(w) = w / (ΣW/M) = M·W/ΣW
  BUG 3 → Eigenvector was normalised to max=1 instead of L2 norm = 1
  BUG 4 → N used len(active) instead of fixed N_PANEL = 178

REQUIREMENTS (run once before this cell):
  !pip install networkx scipy pandas numpy -q

CONFIG — change these two paths to match your Colab session:
  BACI_DIR   = folder containing BACI_HS92_Y{year}_V*.csv files
  OUTPUT_CSV = where to save the final CSV
"""

import os, glob, sys
import numpy as np
import pandas as pd
import networkx as nx
from scipy.sparse import csr_matrix
from scipy.sparse.linalg import eigs
import warnings
warnings.filterwarnings("ignore")

# ── CONFIG ────────────────────────────────────────────────────────────────────
BACI_DIR   = "/content/baci_data"                                 # ← your BACI folder
OUTPUT_CSV = "/content/drive/MyDrive/centrality_measures_1995_2024.csv"  # ← output path
YEARS      = list(range(2005, 2025))                              # 2005 – 2024

# Fixed panel: always 178 countries, used for ALL normalisations
N_PANEL    = 175    # BUG 4 FIX: never use len(active) for normalisation

# ── 178-COUNTRY PANEL ─────────────────────────────────────────────────────────

COUNTRY_PANEL = [
    (  1, 533,"ABW","Aruba"),               (  2,   4,"AFG","Afghanistan"),
    (  3,  24,"AGO","Angola"),              (  4,   8,"ALB","Albania"),
   (  6, 784,"ARE","United Arab Emirates"),
    (  7,  32,"ARG","Argentina"),           (  8,  51,"ARM","Armenia"),
    (  9,  28,"ATG","Antigua and Barbuda"), ( 10,  36,"AUS","Australia"),
    ( 11,  40,"AUT","Austria"),             ( 12,  31,"AZE","Azerbaijan"),
    ( 13, 108,"BDI","Burundi"),             ( 14, 204,"BEN","Benin"),
    ( 15, 854,"BFA","Burkina Faso"),        ( 16,  50,"BGD","Bangladesh"),
    ( 17, 100,"BGR","Bulgaria"),            ( 18,  48,"BHR","Bahrain"),
    ( 19,  70,"BIH","Bosnia and Herzegovina"),( 20, 112,"BLR","Belarus"),
     ( 22,  84,"BLZ","Belize"),
    ( 23,  60,"BMU","Bermuda"),             ( 24,  68,"BOL","Bolivia"),
    ( 25,  76,"BRA","Brazil"),             ( 26,  52,"BRB","Barbados"),
    ( 27,  96,"BRN","Brunei Darussalam"),  ( 28,  64,"BTN","Bhutan"),
    ( 29, 140,"CAF","Central African Republic"),( 30, 124,"CAN","Canada"),
    ( 31, 757,"CHE","Switzerland"),        ( 32, 152,"CHL","Chile"),
    ( 33, 156,"CHN","China"),              ( 34, 384,"CIV","Cote d'Ivoire"),
    ( 35, 120,"CMR","Cameroon"),           ( 36, 178,"COG","Congo"),
    ( 37, 170,"COL","Colombia"),           ( 38, 188,"CRI","Costa Rica"),
    ( 39, 192,"CUB","Cuba"),               ( 40, 136,"CYM","Cayman Islands"),
    ( 41, 196,"CYP","Cyprus"),             ( 42, 203,"CZE","Czech Republic"),
    ( 43, 276,"DEU","Germany"),            ( 44, 212,"DMA","Dominica"),
    ( 45, 208,"DNK","Denmark"),            ( 46, 214,"DOM","Dominican Republic"),
    ( 47,  12,"DZA","Algeria"),            ( 48, 218,"ECU","Ecuador"),
    ( 49, 818,"EGY","Egypt"),              ( 50, 724,"ESP","Spain"),
    ( 51, 233,"EST","Estonia"),            ( 52, 231,"ETH","Ethiopia"),
    ( 53, 246,"FIN","Finland"),            ( 54, 242,"FJI","Fiji"),
    ( 55, 251,"FRA","France"),             ( 56, 583,"FSM","Micronesia"),
    ( 57, 266,"GAB","Gabon"),              ( 58, 826,"GBR","United Kingdom"),
    ( 59, 268,"GEO","Georgia"),            ( 60, 288,"GHA","Ghana"),
    ( 61, 324,"GIN","Guinea"),             ( 62, 270,"GMB","Gambia"),
    ( 63, 624,"GNB","Guinea-Bissau"),      ( 64, 226,"GNQ","Equatorial Guinea"),
    ( 65, 300,"GRC","Greece"),             ( 66, 320,"GTM","Guatemala"),
    ( 67, 328,"GUY","Guyana"),             ( 68, 344,"HKG","Hong Kong"),
    ( 69, 340,"HND","Honduras"),           ( 70, 191,"HRV","Croatia"),
    ( 71, 332,"HTI","Haiti"),              ( 72, 348,"HUN","Hungary"),
    ( 73, 360,"IDN","Indonesia"),          ( 74, 699,"IND","India"),
    ( 75, 372,"IRL","Ireland"),            ( 76, 364,"IRN","Iran"),
    ( 77, 368,"IRQ","Iraq"),               ( 78, 352,"ISL","Iceland"),
    ( 79, 376,"ISR","Israel"),             ( 80, 380,"ITA","Italy"),
    ( 81, 388,"JAM","Jamaica"),            ( 82, 400,"JOR","Jordan"),
    ( 83, 392,"JPN","Japan"),              ( 84, 398,"KAZ","Kazakhstan"),
    ( 85, 404,"KEN","Kenya"),              ( 86, 417,"KGZ","Kyrgyzstan"),
    ( 87, 116,"KHM","Cambodia"),           ( 88, 659,"KNA","Saint Kitts and Nevis"),
    ( 89, 410,"KOR","South Korea"),        ( 90, 414,"KWT","Kuwait"),
    ( 91, 418,"LAO","Laos"),               ( 92, 422,"LBN","Lebanon"),
    ( 93, 430,"LBR","Liberia"),            ( 94, 434,"LBY","Libya"),
    ( 95, 662,"LCA","Saint Lucia"),        ( 96, 144,"LKA","Sri Lanka"),
    ( 97, 440,"LTU","Lithuania"),          ( 98, 428,"LVA","Latvia"),
    ( 99, 446,"MAC","Macau"),              (100, 504,"MAR","Morocco"),
    (101, 498,"MDA","Moldova"),            (102, 450,"MDG","Madagascar"),
    (103, 462,"MDV","Maldives"),           (104, 484,"MEX","Mexico"),
    (105, 584,"MHL","Marshall Islands"),   (106, 807,"MKD","Macedonia"),
    (107, 466,"MLI","Mali"),               (108, 470,"MLT","Malta"),
    (109, 104,"MMR","Myanmar"),            (110, 496,"MNG","Mongolia"),
    (111, 508,"MOZ","Mozambique"),         (112, 478,"MRT","Mauritania"),
    (113, 480,"MUS","Mauritius"),          (114, 454,"MWI","Malawi"),
    (115, 458,"MYS","Malaysia"),           (116, 540,"NCL","New Caledonia"),
    (117, 562,"NER","Niger"),              (118, 566,"NGA","Nigeria"),
    (119, 558,"NIC","Nicaragua"),          (120, 528,"NLD","Netherlands"),
    (121, 579,"NOR","Norway"),             (122, 524,"NPL","Nepal"),
    (123, 554,"NZL","New Zealand"),        (124, 512,"OMN","Oman"),
    (125, 586,"PAK","Pakistan"),           (126, 591,"PAN","Panama"),
    (127, 604,"PER","Peru"),               (128, 608,"PHL","Philippines"),
    (129, 598,"PNG","Papua New Guinea"),   (130, 616,"POL","Poland"),
    (131, 408,"PRK","North Korea"),        (132, 620,"PRT","Portugal"),
    (133, 600,"PRY","Paraguay"),           (134, 634,"QAT","Qatar"),
    (135, 642,"ROM","Romania"),            (136, 643,"RUS","Russian Federation"),
    (137, 646,"RWA","Rwanda"),             (138, 682,"SAU","Saudi Arabia"),
    (140, 686,"SEN","Senegal"),
    (141, 702,"SGP","Singapore"),          (142,  90,"SLB","Solomon Islands"),
    (143, 694,"SLE","Sierra Leone"),       (144, 222,"SLV","El Salvador"),
    (145, 706,"SOM","Somalia"),            (146, 740,"SUR","Suriname"),
    (147, 703,"SVK","Slovak Republic"),    (148, 705,"SVN","Slovenia"),
    (149, 752,"SWE","Sweden"),             (150, 690,"SYC","Seychelles"),
    (151, 760,"SYR","Syria"),              (152, 148,"TCD","Chad"),
    (153, 768,"TGO","Togo"),               (154, 764,"THA","Thailand"),
    (155, 762,"TJK","Tajikistan"),         (156, 795,"TKM","Turkmenistan"),
    (157, 780,"TTO","Trinidad and Tobago"),(158, 788,"TUN","Tunisia"),
    (159, 792,"TUR","Turkey"),             (160, 490,"TWN","Taiwan"),
    (161, 834,"TZA","Tanzania"),           (162, 800,"UGA","Uganda"),
    (163, 804,"UKR","Ukraine"),            (164, 858,"URY","Uruguay"),
    (165, 842,"USA","United States"),      (166, 860,"UZB","Uzbekistan"),
    (167, 670,"VCT","Saint Vincent and the Grenadines"),
    (168, 862,"VEN","Venezuela"),          (169,  92,"VGB","Virgin Islands"),
    (170, 704,"VNM","Viet Nam"),           (171, 548,"VUT","Vanuatu"),
    (172, 882,"WSM","Samoa"),              (173, 887,"YEM","Yemen"),

    (176, 180,"ZAR","Congo Dem. Rep."),
    (177, 894,"ZMB","Zambia"),             (178, 716,"ZWE","Zimbabwe"),
    (179, 56, "BEL","Belgium"),            (180, 442, "LUX","Luxembourg")
]
"""COUNTRY_PANEL = [
    (  1, 533,"ABW","Aruba"),               (  2,   4,"AFG","Afghanistan"),
    (  3,  24,"AGO","Angola"),              (  4,   8,"ALB","Albania"),
    (  5, 530,"ANT","Netherlands Antilles"),(  6, 784,"ARE","United Arab Emirates"),
    (  7,  32,"ARG","Argentina"),           (  8,  51,"ARM","Armenia"),
    (  9,  28,"ATG","Antigua and Barbuda"), ( 10,  36,"AUS","Australia"),
    ( 11,  40,"AUT","Austria"),             ( 12,  31,"AZE","Azerbaijan"),
    ( 13, 108,"BDI","Burundi"),             ( 14, 204,"BEN","Benin"),
    ( 15, 854,"BFA","Burkina Faso"),        ( 16,  50,"BGD","Bangladesh"),
    ( 17, 100,"BGR","Bulgaria"),            ( 18,  48,"BHR","Bahrain"),
    ( 19,  70,"BIH","Bosnia and Herzegovina"),( 20, 112,"BLR","Belarus"),
    ( 21,  58,"BLX","Belgium-Luxembourg"), ( 22,  84,"BLZ","Belize"),
    ( 23,  60,"BMU","Bermuda"),             ( 24,  68,"BOL","Bolivia"),
    ( 25,  76,"BRA","Brazil"),             ( 26,  52,"BRB","Barbados"),
    ( 27,  96,"BRN","Brunei Darussalam"),  ( 28,  64,"BTN","Bhutan"),
    ( 29, 140,"CAF","Central African Republic"),( 30, 124,"CAN","Canada"),
    ( 31, 757,"CHE","Switzerland"),        ( 32, 152,"CHL","Chile"),
    ( 33, 156,"CHN","China"),              ( 34, 384,"CIV","Cote d'Ivoire"),
    ( 35, 120,"CMR","Cameroon"),           ( 36, 178,"COG","Congo"),
    ( 37, 170,"COL","Colombia"),           ( 38, 188,"CRI","Costa Rica"),
    ( 39, 192,"CUB","Cuba"),               ( 40, 136,"CYM","Cayman Islands"),
    ( 41, 196,"CYP","Cyprus"),             ( 42, 203,"CZE","Czech Republic"),
    ( 43, 276,"DEU","Germany"),            ( 44, 212,"DMA","Dominica"),
    ( 45, 208,"DNK","Denmark"),            ( 46, 214,"DOM","Dominican Republic"),
    ( 47,  12,"DZA","Algeria"),            ( 48, 218,"ECU","Ecuador"),
    ( 49, 818,"EGY","Egypt"),              ( 50, 724,"ESP","Spain"),
    ( 51, 233,"EST","Estonia"),            ( 52, 231,"ETH","Ethiopia"),
    ( 53, 246,"FIN","Finland"),            ( 54, 242,"FJI","Fiji"),
    ( 55, 251,"FRA","France"),             ( 56, 583,"FSM","Micronesia"),
    ( 57, 266,"GAB","Gabon"),              ( 58, 826,"GBR","United Kingdom"),
    ( 59, 268,"GEO","Georgia"),            ( 60, 288,"GHA","Ghana"),
    ( 61, 324,"GIN","Guinea"),             ( 62, 270,"GMB","Gambia"),
    ( 63, 624,"GNB","Guinea-Bissau"),      ( 64, 226,"GNQ","Equatorial Guinea"),
    ( 65, 300,"GRC","Greece"),             ( 66, 320,"GTM","Guatemala"),
    ( 67, 328,"GUY","Guyana"),             ( 68, 344,"HKG","Hong Kong"),
    ( 69, 340,"HND","Honduras"),           ( 70, 191,"HRV","Croatia"),
    ( 71, 332,"HTI","Haiti"),              ( 72, 348,"HUN","Hungary"),
    ( 73, 360,"IDN","Indonesia"),          ( 74, 699,"IND","India"),
    ( 75, 372,"IRL","Ireland"),            ( 76, 364,"IRN","Iran"),
    ( 77, 368,"IRQ","Iraq"),               ( 78, 352,"ISL","Iceland"),
    ( 79, 376,"ISR","Israel"),             ( 80, 380,"ITA","Italy"),
    ( 81, 388,"JAM","Jamaica"),            ( 82, 400,"JOR","Jordan"),
    ( 83, 392,"JPN","Japan"),              ( 84, 398,"KAZ","Kazakhstan"),
    ( 85, 404,"KEN","Kenya"),              ( 86, 417,"KGZ","Kyrgyzstan"),
    ( 87, 116,"KHM","Cambodia"),           ( 88, 659,"KNA","Saint Kitts and Nevis"),
    ( 89, 410,"KOR","South Korea"),        ( 90, 414,"KWT","Kuwait"),
    ( 91, 418,"LAO","Laos"),               ( 92, 422,"LBN","Lebanon"),
    ( 93, 430,"LBR","Liberia"),            ( 94, 434,"LBY","Libya"),
    ( 95, 662,"LCA","Saint Lucia"),        ( 96, 144,"LKA","Sri Lanka"),
    ( 97, 440,"LTU","Lithuania"),          ( 98, 428,"LVA","Latvia"),
    ( 99, 446,"MAC","Macau"),              (100, 504,"MAR","Morocco"),
    (101, 498,"MDA","Moldova"),            (102, 450,"MDG","Madagascar"),
    (103, 462,"MDV","Maldives"),           (104, 484,"MEX","Mexico"),
    (105, 584,"MHL","Marshall Islands"),   (106, 807,"MKD","Macedonia"),
    (107, 466,"MLI","Mali"),               (108, 470,"MLT","Malta"),
    (109, 104,"MMR","Myanmar"),            (110, 496,"MNG","Mongolia"),
    (111, 508,"MOZ","Mozambique"),         (112, 478,"MRT","Mauritania"),
    (113, 480,"MUS","Mauritius"),          (114, 454,"MWI","Malawi"),
    (115, 458,"MYS","Malaysia"),           (116, 540,"NCL","New Caledonia"),
    (117, 562,"NER","Niger"),              (118, 566,"NGA","Nigeria"),
    (119, 558,"NIC","Nicaragua"),          (120, 528,"NLD","Netherlands"),
    (121, 579,"NOR","Norway"),             (122, 524,"NPL","Nepal"),
    (123, 554,"NZL","New Zealand"),        (124, 512,"OMN","Oman"),
    (125, 586,"PAK","Pakistan"),           (126, 591,"PAN","Panama"),
    (127, 604,"PER","Peru"),               (128, 608,"PHL","Philippines"),
    (129, 598,"PNG","Papua New Guinea"),   (130, 616,"POL","Poland"),
    (131, 408,"PRK","North Korea"),        (132, 620,"PRT","Portugal"),
    (133, 600,"PRY","Paraguay"),           (134, 634,"QAT","Qatar"),
    (135, 642,"ROM","Romania"),            (136, 643,"RUS","Russian Federation"),
    (137, 646,"RWA","Rwanda"),             (138, 682,"SAU","Saudi Arabia"),
    (139, 736,"SDN","Sudan"),              (140, 686,"SEN","Senegal"),
    (141, 702,"SGP","Singapore"),          (142,  90,"SLB","Solomon Islands"),
    (143, 694,"SLE","Sierra Leone"),       (144, 222,"SLV","El Salvador"),
    (145, 706,"SOM","Somalia"),            (146, 740,"SUR","Suriname"),
    (147, 703,"SVK","Slovak Republic"),    (148, 705,"SVN","Slovenia"),
    (149, 752,"SWE","Sweden"),             (150, 690,"SYC","Seychelles"),
    (151, 760,"SYR","Syria"),              (152, 148,"TCD","Chad"),
    (153, 768,"TGO","Togo"),               (154, 764,"THA","Thailand"),
    (155, 762,"TJK","Tajikistan"),         (156, 795,"TKM","Turkmenistan"),
    (157, 780,"TTO","Trinidad and Tobago"),(158, 788,"TUN","Tunisia"),
    (159, 792,"TUR","Turkey"),             (160, 490,"TWN","Taiwan"),
    (161, 834,"TZA","Tanzania"),           (162, 800,"UGA","Uganda"),
    (163, 804,"UKR","Ukraine"),            (164, 858,"URY","Uruguay"),
    (165, 842,"USA","United States"),      (166, 860,"UZB","Uzbekistan"),
    (167, 670,"VCT","Saint Vincent and the Grenadines"),
    (168, 862,"VEN","Venezuela"),          (169,  92,"VGB","Virgin Islands"),
    (170, 704,"VNM","Viet Nam"),           (171, 548,"VUT","Vanuatu"),
    (172, 882,"WSM","Samoa"),              (173, 887,"YEM","Yemen"),
    (174, 891,"YUG","Serbia and Montenegro"),
    (175, 711,"ZAF","South Africa"),       (176, 180,"ZAR","Congo Dem. Rep."),
    (177, 894,"ZMB","Zambia"),             (178, 716,"ZWE","Zimbabwe"),
]"""
"""COUNTRY_PANEL = [
    ( 7,  32,"ARG","Argentina"),
    ( 10,  36,"AUS","Australia"),            ( 11,  40,"AUT","Austria"),
    ( 16,  50,"BGD","Bangladesh"),           ( 17, 100,"BGR","Bulgaria"),
    ( 21,  56,"BEL","Belgium"),              ( 25,  76,"BRA","Brazil"),
    ( 30, 124,"CAN","Canada"),               ( 31, 757,"CHE","Switzerland"),
    ( 33, 156,"CHN","China"),
    ( 41, 196,"CYP","Cyprus"),               ( 42, 203,"CZE","Czech Republic"),
    ( 43, 276,"DEU","Germany"),              ( 45, 208,"DNK","Denmark"),
    ( 50, 724,"ESP","Spain"),                ( 51, 233,"EST","Estonia"),
    ( 53, 246,"FIN","Finland"),              ( 55, 251,"FRA","France"),
    ( 58, 826,"GBR","United Kingdom"),       ( 65, 300,"GRC","Greece"),
    ( 70, 191,"HRV","Croatia"),              ( 72, 348,"HUN","Hungary"),
    ( 73, 360,"IDN","Indonesia"),            ( 74, 699,"IND","India"),
    ( 75, 372,"IRL","Ireland"),              ( 80, 380,"ITA","Italy"),
    ( 83, 392,"JPN","Japan"),                ( 84, 398,"KAZ","Kazakhstan"),
    ( 89, 410,"KOR","South Korea"),          ( 91, 418,"LAO","Laos"),
    ( 96, 144,"LKA","Sri Lanka"),            ( 97, 440,"LTU","Lithuania"),
    ( 98, 428,"LVA","Latvia"),               (103, 462,"MDV","Maldives"),
    (104, 484,"MEX","Mexico"),
    (108, 470,"MLT","Malta"),                (115, 458,"MYS","Malaysia"),
    (120, 528,"NLD","Netherlands"),          (121, 579,"NOR","Norway"),
    (122, 524,"NPL","Nepal"),                (123, 554,"NZL","New Zealand"),
    (125, 586,"PAK","Pakistan"),
    (128, 608,"PHL","Philippines"),          (130, 616,"POL","Poland"),
    (132, 620,"PRT","Portugal"),             (135, 642,"ROM","Romania"),
    (136, 643,"RUS","Russian Federation"),   (141, 702,"SGP","Singapore"),
    (147, 703,"SVK","Slovak Republic"),      (148, 705,"SVN","Slovenia"),
    (149, 752,"SWE","Sweden"),               (154, 764,"THA","Thailand"),
    (159, 792,"TUR","Turkey"),
    (165, 842,"USA","United States"),        (170, 704,"VNM","Viet Nam"),
    (171, 442,"LUX","Luxembourg"),
]
COUNTRY_PANEL = [
    ( 10,  36,"AUS","Australia"),            ( 11,  40,"AUT","Austria"),
    ( 16,  50,"BGD","Bangladesh"),           ( 17, 100,"BGR","Bulgaria"),
    ( 21,  56,"BEL","Belgium"),              ( 25,  76,"BRA","Brazil"),
    ( 30, 124,"CAN","Canada"),               ( 31, 757,"CHE","Switzerland"),
    ( 33, 156,"CHN","China"),
    ( 41, 196,"CYP","Cyprus"),               ( 42, 203,"CZE","Czech Republic"),
    ( 43, 276,"DEU","Germany"),              ( 45, 208,"DNK","Denmark"),
    ( 50, 724,"ESP","Spain"),                ( 51, 233,"EST","Estonia"),
    ( 53, 246,"FIN","Finland"),              ( 55, 251,"FRA","France"),
    ( 58, 826,"GBR","United Kingdom"),       ( 65, 300,"GRC","Greece"),
    ( 70, 191,"HRV","Croatia"),              ( 72, 348,"HUN","Hungary"),
    ( 73, 360,"IDN","Indonesia"),            ( 74, 699,"IND","India"),
    ( 75, 372,"IRL","Ireland"),              ( 80, 380,"ITA","Italy"),
    ( 83, 392,"JPN","Japan"),                ( 84, 398,"KAZ","Kazakhstan"),
    ( 89, 410,"KOR","South Korea"),          ( 91, 418,"LAO","Laos"),
    ( 96, 144,"LKA","Sri Lanka"),            ( 97, 440,"LTU","Lithuania"),
    ( 98, 428,"LVA","Latvia"),               (103, 462,"MDV","Maldives"),
    (104, 484,"MEX","Mexico"),
    (108, 470,"MLT","Malta"),                (115, 458,"MYS","Malaysia"),
    (120, 528,"NLD","Netherlands"),          (121, 579,"NOR","Norway"),
    (122, 524,"NPL","Nepal"),
    (125, 586,"PAK","Pakistan"),
    (128, 608,"PHL","Philippines"),          (130, 616,"POL","Poland"),
    (132, 620,"PRT","Portugal"),             (135, 642,"ROM","Romania"),
    (136, 643,"RUS","Russian Federation"),   (141, 702,"SGP","Singapore"),
    (147, 703,"SVK","Slovak Republic"),      (148, 705,"SVN","Slovenia"),
    (149, 752,"SWE","Sweden"),               (154, 764,"THA","Thailand"),
    (159, 792,"TUR","Turkey"),
    (165, 842,"USA","United States"),        (170, 704,"VNM","Viet Nam"),
    (171, 442,"LUX","Luxembourg"),
]"""

NUMERIC_TO_ISO3 = {row[1]: row[2] for row in COUNTRY_PANEL}


ISO3_TO_ID   = {row[2]: row[0] for row in COUNTRY_PANEL}
PANEL_ISO3   = [row[2] for row in COUNTRY_PANEL]

COL_ORDER = [
    "t","id","i","iso3","country",
    "Out-degree","In-degree","Out-strength","In-strength",
    "Out-degree_percent","In-degree percent",
    "Out-strength-percent","In-strength_percent",
    "Out-closenness","In-closenness",
    "W-Out-closenness","W-In-closenness",
    "Out-eigenvector","In-eigenvector",
    "W-Out-eigenvector","W-In-eigenvector",
]


# ── STEP 1: LOAD BACI ─────────────────────────────────────────────────────────
def load_baci_year(year):
    pattern = os.path.join(BACI_DIR, f"BACI_HS92_Y{year}_V*.csv")
    files   = glob.glob(pattern)
    if not files:
        raise FileNotFoundError(f"No BACI file for year {year} — pattern: {pattern}")

    print(f"  Loading {os.path.basename(files[0])} ...", end=" ", flush=True)
    df = pd.read_csv(files[0], dtype={"k": str, "i": int, "j": int, "v": float})
    print(f"{len(df):,} product-level rows")

    # Map ISO numeric → ISO3, aggregate across products
    df["exp"] = df["i"].map(lambda x: NUMERIC_TO_ISO3.get(x))
    df["imp"] = df["j"].map(lambda x: NUMERIC_TO_ISO3.get(x))
    df = df.dropna(subset=["exp","imp"])
    df = df[df["exp"] != df["imp"]]   # drop self-loops

    bilateral = (df.groupby(["exp","imp"])["v"]
                   .sum().reset_index()
                   .rename(columns={"v":"val"}))
    print(f"  → {len(bilateral):,} bilateral flows | "
          f"{bilateral['exp'].nunique()} exporters | "
          f"{bilateral['imp'].nunique()} importers")
    return bilateral


# ── STEP 2: BUILD DIRECTED WEIGHTED GRAPH ────────────────────────────────────
def build_graph(bilateral):
    G = nx.DiGraph()
    G.add_nodes_from(PANEL_ISO3)    # always 178 nodes
    for _, row in bilateral.iterrows():
        if row["exp"] in ISO3_TO_ID and row["imp"] in ISO3_TO_ID:
            G.add_edge(row["exp"], row["imp"], weight=float(row["val"]))
    return G


# ── STEP 3a: DEGREE + STRENGTH ────────────────────────────────────────────────

def compute_degree_strength(G, active):
    M           = G.number_of_edges()
    total_trade = sum(d["weight"] for _,_,d in G.edges(data=True))
    N1          = N_PANEL - 1

    out = {}
    for n in active:
        ol  = G.out_degree(n)
        il  = G.in_degree(n)
        os_ = sum(d["weight"] for _,_,d in G.out_edges(n, data=True))
        is_ = sum(d["weight"] for _,_,d in G.in_edges(n,  data=True))
        out[n] = {
            # Equations 2-3: normalise by (N_PANEL-1)
            "Out-degree":           ol  / N1,
            "In-degree":            il  / N1,
            # BUG 1 FIX: divide by N_PANEL-1, not skip it
            "Out-strength":         os_ / N1,
            "In-strength":          is_ / N1,
            # Equations 4 & 6: percentage of total links / total trade
            "Out-degree_percent":   ol  / M           * 100 if M  > 0 else 0,
            "In-degree percent":    il  / M           * 100 if M  > 0 else 0,
            "Out-strength-percent": os_ / total_trade * 100 if total_trade > 0 else 0,
            "In-strength_percent":  is_ / total_trade * 100 if total_trade > 0 else 0,
        }
    return out


# ── STEP 3b: UNWEIGHTED CLOSENESS ────────────────────────────────────────────

def compute_closeness_unweighted(G, active):
    G_rev = G.reverse(copy=False)
    N1    = N_PANEL - 1

    out = {}
    for n in active:
        # Out-closeness: shortest paths FROM n to all others
        sp  = nx.single_source_shortest_path_length(G, n)
        d   = sum(sp.get(j, N_PANEL) for j in active if j != n)
        out_c = N1 / d  if d > 0 else 0.0

        # In-closeness: shortest paths TO n (= out-closeness on reversed graph)
        spr = nx.single_source_shortest_path_length(G_rev, n)
        dr  = sum(spr.get(j, N_PANEL) for j in active if j != n)
        in_c  = N1 / dr if dr > 0 else 0.0

        out[n] = {"Out-closenness": out_c, "In-closenness": in_c}
    return out


# ── STEP 3c: WEIGHTED CLOSENESS (Opsahl/tnet method) ────────────────────────
# Paper eq. 9-10 + Stata appendix:
#   summ w → r(mean) = ΣW / M
#   omega_ij = W_ij / mean(W) = M * W_ij / ΣW
#   distance_ij = 1 / omega_ij = ΣW / (M * W_ij)
#   C_NW_C = (N-1) / Σ_j l(i→j)   via Dijkstra

def compute_closeness_weighted(G, active):
    M         = G.number_of_edges()
    total_W   = sum(d["weight"] for _,_,d in G.edges(data=True))
    N1        = N_PANEL - 1

    if total_W == 0 or M == 0:
        return {n: {"W-Out-closenness": 0.0, "W-In-closenness": 0.0} for n in active}

    # Build distance graph: edge weight = 1/omega = ΣW / (M * W_ij)
    # (equivalent to: w_average = ΣW/M;  inv_w = w / w_average;  dist = 1/inv_w)
    Gd = nx.DiGraph()
    Gd.add_nodes_from(active)
    for u, v, data in G.edges(data=True):
        w     = data["weight"]
        omega = (M * w) / total_W
        dist  = 1.0 / omega if omega > 0 else 1e12
        Gd.add_edge(u, v, weight=dist)
    Gd_rev = Gd.reverse(copy=False)

    out = {}
    for n in active:
        # Out-closeness weighted
        try:
            sp = dict(nx.single_source_dijkstra_path_length(Gd, n, weight="weight"))
            d  = sum(sp.get(j, 1e12) for j in active if j != n)
            wc_out = N1 / d if d > 0 else 0.0
        except Exception:
            wc_out = 0.0

        # In-closeness weighted (reversed graph)
        try:
            sp = dict(nx.single_source_dijkstra_path_length(Gd_rev, n, weight="weight"))
            d  = sum(sp.get(j, 1e12) for j in active if j != n)
            wc_in  = N1 / d if d > 0 else 0.0
        except Exception:
            wc_in  = 0.0

        out[n] = {"W-Out-closenness": wc_out, "W-In-closenness": wc_in}
    return out


# ── STEP 3d: EIGENVECTOR CENTRALITY ──────────────────────────────────────────

# Out-eigenvector: leading right eigenvector of adjacency matrix A
# In-eigenvector:  leading right eigenvector of Aᵀ  (= left eigenvector of A)
def _leading_eigvec_l2(mat, n):
    """Leading eigenvector of sparse matrix, L2-normalised."""
    try:
        _, vecs = eigs(mat.astype(float), k=1, which="LM", maxiter=10_000, tol=1e-8)
        v    = np.abs(vecs[:, 0].real)
        norm = np.linalg.norm(v)          # BUG 3 FIX: L2 norm, not max
        return v / norm if norm > 1e-12 else np.zeros(n)
    except Exception:
        return np.zeros(n)


def compute_eigenvector(G, active, weighted=False):
    n   = len(active)
    idx = {nd: k for k, nd in enumerate(active)}

    rows, cols, vals = [], [], []
    for u, v, data in G.edges(data=True):
        if u in idx and v in idx:
            rows.append(idx[u])
            cols.append(idx[v])
            vals.append(data["weight"] if weighted else 1.0)

    if not vals:
        zeros = {nd: 0.0 for nd in active}
        return zeros, zeros

    A = csr_matrix((vals, (rows, cols)), shape=(n, n), dtype=float)

    ev_out_vec = _leading_eigvec_l2(A,   n)
    ev_in_vec  = _leading_eigvec_l2(A.T, n)

    ev_out = {nd: float(ev_out_vec[idx[nd]]) for nd in active}
    ev_in  = {nd: float(ev_in_vec[idx[nd]])  for nd in active}
    return ev_out, ev_in


# ── STEP 4: COMBINE INTO DATAFRAME FOR ONE YEAR ───────────────────────────────
def compute_year(year):
    bilateral = load_baci_year(year)
    G = build_graph(bilateral)

    active = [iso for iso in PANEL_ISO3
              if G.out_degree(iso) > 0 or G.in_degree(iso) > 0]
    M = G.number_of_edges()
    print(f"  Active countries: {len(active)} / {N_PANEL} | Edges: {M:,}")

    print("  [1/4] Degree + Strength ...",        end=" ", flush=True)
    deg = compute_degree_strength(G, active)

    print("done\n  [2/4] Unweighted closeness ...", end=" ", flush=True)
    cl_u = compute_closeness_unweighted(G, active)

    print("done\n  [3/4] Weighted closeness ...",   end=" ", flush=True)
    cl_w = compute_closeness_weighted(G, active)

    print("done\n  [4/4] Eigenvector centrality ...", end=" ", flush=True)
    ev_u_out, ev_u_in = compute_eigenvector(G, active, weighted=False)
    ev_w_out, ev_w_in = compute_eigenvector(G, active, weighted=True)
    print("done")

    # Build one row per panel country (inactive countries → all zeros)
    records = []
    for cid, cnum, iso3, cname in COUNTRY_PANEL:
        row = {"t": year, "id": cid, "i": cnum, "iso3": iso3, "country": cname}
        if iso3 in active:
            row.update(deg[iso3])
            row.update(cl_u[iso3])
            row.update(cl_w[iso3])
            row["Out-eigenvector"]   = ev_u_out[iso3]
            row["In-eigenvector"]    = ev_u_in[iso3]
            row["W-Out-eigenvector"] = ev_w_out[iso3]
            row["W-In-eigenvector"]  = ev_w_in[iso3]
        else:
            # Country not trading this year: set all measures to 0
            for col in COL_ORDER[5:]:
                row[col] = 0.0
        records.append(row)

    return pd.DataFrame(records)[COL_ORDER]


# ── STEP 5: MAIN LOOP ─────────────────────────────────────────────────────────
print("=" * 60)
print(f"  World Trade Network Centrality — {YEARS[0]} to {YEARS[-1]}")
print(f"  Panel: {N_PANEL} countries | Output: {OUTPUT_CSV}")
print("=" * 60)

all_dfs = []
skipped = []

for year in YEARS:
    print(f"\n{'─'*60}")
    print(f"  YEAR {year}")
    print(f"{'─'*60}")
    try:
        df_year = compute_year(year)
        all_dfs.append(df_year)
        print(f"  ✓  Year {year} complete")
    except Exception as e:
        print(f"  ✗  Year {year} SKIPPED: {e}")
        skipped.append(year)

# ── STEP 6: SAVE ─────────────────────────────────────────────────────────────
if all_dfs:
    final = pd.concat(all_dfs, ignore_index=True).sort_values(["t","id"])
    final.to_csv(OUTPUT_CSV, sep=";", index=False, float_format="%.7g")
    print(f"\n{'='*60}")
    print(f"  SAVED: {OUTPUT_CSV}")
    print(f"  Rows  : {len(final):,}  ({len(YEARS)} years × {N_PANEL} countries)")
    print(f"  Years : {sorted(final['t'].unique())}")
    if skipped:
        print(f"  Skipped years: {skipped}")
    print("=" * 60)
else:
    print("\n  ERROR: Nothing computed. Check BACI_DIR path.")


# ── OPTIONAL: QUICK VALIDATION AGAINST ORIGINAL CSV ──────────────────────────
# Set ORIG_CSV to your original 1995-2010 file path and run this block
# to verify your 2007 results are close to the original.
#
# ORIG_CSV = "/content/drive/MyDrive/Cepii_centrality_measures.csv"
#
# orig = pd.read_csv(ORIG_CSV, sep=";")
# orig07 = orig[orig['t']==2007].set_index('iso3')
# new07  = final[final['t']==2007].set_index('iso3')
# common = orig07.index.intersection(new07.index)
# check_cols = ["Out-degree","Out-strength","Out-closenness","W-Out-closenness",
#               "Out-eigenvector","W-Out-eigenvector"]
# print("\n── Validation: 2007 mean-absolute-difference ──")
# for col in check_cols:
#     diff = (orig07.loc[common,col].astype(float)
#             - new07.loc[common,col].astype(float)).abs()
#     pct  = (diff / (orig07.loc[common,col].abs()+1e-10)).mean()*100
#     print(f"  {col:<35} mean_abs={diff.mean():.5f}  mean_pct≈{pct:.1f}%")
# print("  (Remaining % gap = BACI data revision V202601 vs 2013 version — expected)")
```
