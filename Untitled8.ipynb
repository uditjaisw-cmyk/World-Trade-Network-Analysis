```python
"""
==============================================================
Semiconductor Trade Network — Centrality Measures (2017-2024)
==============================================================
Uses BACI HS17 data filtered to 59 semiconductor-related
HS product codes from your Excel file.

Output CSV columns (semicolon-separated):
  t, id, i, iso3, country,
  Out-degree, In-degree,
  Out-strength, In-strength,
  Out-degree_percent, In-degree percent,
  Out-strength-percent, In-strength_percent,
  Out-closenness, In-closenness,
  W-Out-closenness, W-In-closenness,
  Out-eigenvector, In-eigenvector,
  W-Out-eigenvector, W-In-eigenvector

HOW TO USE IN GOOGLE COLAB:
  Cell 1: Mount Drive + unzip BACI_HS17_V202601.zip into /content/baci_hs17/
  Cell 2: !pip install networkx scipy pandas numpy openpyxl -q
  Cell 3: Paste this entire file and run

DATA SOURCE:
  https://www.cepii.fr/DATA_DOWNLOAD/baci/data/BACI_HS17_V202601.zip
  Files inside: BACI_HS17_Y2017_V202601.csv ... BACI_HS17_Y2024_V202601.csv

FORMULAS (exactly matching CEPII original, all verified):
  Out-degree           = out_links  / (N-1)
  Out-strength         = Σ W_ij     / (N-1)       ← avg, not raw sum
  Out-degree_percent   = out_links  / M  * 100
  Out-strength-percent = Σ W_ij     / ΣW * 100
  Out-closenness       = (N-1) / Σ geodesic distances
  W-Out-closenness     = (N-1) / Σ dijkstra(1/omega), omega=N·W/ΣW
  Out-eigenvector      = leading eigenvec of A, L2-normalised (‖v‖₂=1)
  W-Out-eigenvector    = leading eigenvec of W, L2-normalised
==============================================================
"""

import os, glob, time
import numpy as np
import pandas as pd
import networkx as nx
from scipy.sparse import csr_matrix
from scipy.sparse.linalg import eigs
import warnings
warnings.filterwarnings("ignore")

# ══════════════════════════════════════════════════════════════
# CONFIG  ← change these paths to match your Colab session
# ══════════════════════════════════════════════════════════════
BACI_DIR   = "/content/baci_hs17"
OUTPUT_CSV = "/content/drive/MyDrive/semiconductor_centrality_HS17_2017_2024.csv"
YEARS      = list(range(2017, 2025))   # 2017–2024

# ══════════════════════════════════════════════════════════════
# 59 SEMICONDUCTOR HS17 PRODUCT CODES (from your Excel file)
# ══════════════════════════════════════════════════════════════
SEMICON_CODES = {
    "252800",  # Natural borates
    "280421",  # Rare gases: argon
    "280429",  # Rare gases: other
    "280461",  # Silicon ≥99.99%
    "280480",  # Arsenic
    "281000",  # Oxides of boron / boric acids
    "281212",  # Phosphorus oxychloride
    "282560",  # Germanium oxides and zirconium dioxide
    "283325",  # Sulphates of copper
    "284011",  # Borates: disodium tetraborate, anhydrous
    "284019",  # Borates: disodium tetraborate, other
    "284920",  # Carbides of silicon
    "285000",  # Hydrides, nitrides, azides, silicides and borides
    "370130",  # Photographic plates/film, flat, >225mm
    "370199",  # Photographic plates/film, flat, ≤255mm
    "370790",  # Photographic goods: chemical preparations
    "381800",  # Chemical elements doped for electronics (wafers/discs)
    "811292",  # Gallium, germanium, hafnium, indium... unwrought/powders
    "811299",  # Gallium, germanium, hafnium, indium... other articles
    "841459",  # Fans n.e.c. 8414.51
    "841950",  # Heat exchange units
    "842129",  # Machinery: filtering/purifying liquids n.e.c.
    "842139",  # Machinery: filtering/purifying gases
    "842199",  # Parts for filtering/purifying machinery
    "848610",  # Machines for manufacturing semiconductor boules/wafers
    "848620",  # Machines for manufacturing semiconductor devices/ICs
    "848630",  # Machines for manufacturing flat panel displays
    "848640",  # Machines for masks/reticles/assembling semiconductor devices
    "848690",  # Parts for 8486 machines
    "852351",  # Solid-state non-volatile storage devices
    "852352",  # Smart cards
    "852359",  # Semiconductor media: other
    "853290",  # Parts of electrical capacitors
    "853390",  # Parts of resistors
    "854110",  # Diodes (other than photosensitive/LED)
    "854121",  # Transistors, dissipation <1W
    "854129",  # Transistors, dissipation ≥1W
    "854130",  # Thyristors, diacs and triacs
    "854140",  # Photosensitive semiconductor devices / photovoltaic cells
    "854150",  # Photosensitive semiconductor devices n.e.c.
    "854160",  # Mounted piezo-electric crystals
    "854190",  # Parts for diodes, transistors, similar semiconductor devices
    "854231",  # Electronic ICs: processors and controllers
    "854232",  # Electronic ICs: memories
    "854233",  # Electronic ICs: amplifiers
    "854239",  # Electronic ICs: n.e.c.
    "854290",  # Parts of electronic integrated circuits
    "900120",  # Optical elements: polarising material
    "900190",  # Optical elements: lenses, prisms, mirrors n.e.c.
    "900219",  # Objective lenses (not cameras/projectors)
    "900220",  # Mounted optical filters
    "900290",  # Optical elements n.e.c. mounted
    "900699",  # Photographic flashlight apparatus parts
    "901210",  # Microscopes excl. optical: diffraction apparatus
    "901290",  # Parts for diffraction apparatus
    "903082",  # Instruments for measuring semiconductor wafers/devices
    "903084",  # Instruments n.e.c. with recording device
    "903141",  # Optical instruments for inspecting semiconductor wafers
    "903300",  # Parts for chapter 90 machines/instruments
}

print(f"Semiconductor product codes loaded: {len(SEMICON_CODES)}")

# ══════════════════════════════════════════════════════════════
# COUNTRY PANEL (same 178 countries as CEPII original)
# ══════════════════════════════════════════════════════════════
"""COUNTRY_PANEL = [
    (  1, 533,"ABW","Aruba"),
    (  2,   4,"AFG","Afghanistan"),
    (  3,  24,"AGO","Angola"),
    (  4,   8,"ALB","Albania"),
    (  5, 530,"ANT","Netherlands antilles"),
    (  6, 784,"ARE","United arab emirates"),
    (  7,  32,"ARG","Argentina"),
    (  8,  51,"ARM","Armenia"),
    (  9,  28,"ATG","Antigua and barbuda"),
    ( 10,  36,"AUS","Australia"),
    ( 11,  40,"AUT","Austria"),
    ( 12,  31,"AZE","Azerbaijan"),
    ( 13, 108,"BDI","Burundi"),
    ( 14, 204,"BEN","Benin"),
    ( 15, 854,"BFA","Burkina faso"),
    ( 16,  50,"BGD","Bangladesh"),
    ( 17, 100,"BGR","Bulgaria"),
    ( 18,  48,"BHR","Bahrain"),
    ( 19,  70,"BIH","Bosnia and herzegowina"),
    ( 20, 112,"BLR","Belarus"),
    ( 21,  58,"BLX","Belgium-Luxembourg"),
    ( 22,  84,"BLZ","Belize"),
    ( 23,  60,"BMU","Bermuda"),
    ( 24,  68,"BOL","Bolivia"),
    ( 25,  76,"BRA","Brazil"),
    ( 26,  52,"BRB","Barbados"),
    ( 27,  96,"BRN","Brunei darussalam"),
    ( 28,  64,"BTN","Bhutan"),
    ( 29, 140,"CAF","Central african republic"),
    ( 30, 124,"CAN","Canada"),
    ( 31, 757,"CHE","Switzerland"),
    ( 32, 152,"CHL","Chile"),
    ( 33, 156,"CHN","China"),
    ( 34, 384,"CIV","Cote d'ivoire"),
    ( 35, 120,"CMR","Cameroon"),
    ( 36, 178,"COG","Congo"),
    ( 37, 170,"COL","Colombia"),
    ( 38, 188,"CRI","Costa rica"),
    ( 39, 192,"CUB","Cuba"),
    ( 40, 136,"CYM","Cayman islands"),
    ( 41, 196,"CYP","Cyprus"),
    ( 42, 203,"CZE","Czech republic"),
    ( 43, 276,"DEU","Germany"),
    ( 44, 212,"DMA","Dominica"),
    ( 45, 208,"DNK","Denmark"),
    ( 46, 214,"DOM","Dominican republic"),
    ( 47,  12,"DZA","Algeria"),
    ( 48, 218,"ECU","Ecuador"),
    ( 49, 818,"EGY","Egypt"),
    ( 50, 724,"ESP","Spain"),
    ( 51, 233,"EST","Estonia"),
    ( 52, 231,"ETH","Ethiopia"),
    ( 53, 246,"FIN","Finland"),
    ( 54, 242,"FJI","Fiji"),
    ( 55, 251,"FRA","France"),
    ( 56, 583,"FSM","Micronesia"),
    ( 57, 266,"GAB","Gabon"),
    ( 58, 826,"GBR","United kingdom"),
    ( 59, 268,"GEO","Georgia"),
    ( 60, 288,"GHA","Ghana"),
    ( 61, 324,"GIN","Guinea"),
    ( 62, 270,"GMB","Gambia"),
    ( 63, 624,"GNB","Guinea-bissau"),
    ( 64, 226,"GNQ","Equatorial guinea"),
    ( 65, 300,"GRC","Greece"),
    ( 66, 320,"GTM","Guatemala"),
    ( 67, 328,"GUY","Guyana"),
    ( 68, 344,"HKG","Hong kong"),
    ( 69, 340,"HND","Honduras"),
    ( 70, 191,"HRV","Croatia"),
    ( 71, 332,"HTI","Haiti"),
    ( 72, 348,"HUN","Hungary"),
    ( 73, 360,"IDN","Indonesia"),
    ( 74, 699,"IND","India"),
    ( 75, 372,"IRL","Ireland"),
    ( 76, 364,"IRN","Iran"),
    ( 77, 368,"IRQ","Iraq"),
    ( 78, 352,"ISL","Iceland"),
    ( 79, 376,"ISR","Israel"),
    ( 80, 381,"ITA","Italy"),
    ( 81, 388,"JAM","Jamaica"),
    ( 82, 400,"JOR","Jordan"),
    ( 83, 392,"JPN","Japan"),
    ( 84, 398,"KAZ","Kazakhstan"),
    ( 85, 404,"KEN","Kenya"),
    ( 86, 417,"KGZ","Kyrgyzstan"),
    ( 87, 116,"KHM","Cambodia"),
    ( 88, 659,"KNA","Saint kitts and nevis"),
    ( 89, 410,"KOR","South Korea"),
    ( 90, 414,"KWT","Kuwait"),
    ( 91, 418,"LAO","Laos"),
    ( 92, 422,"LBN","Lebanon"),
    ( 93, 430,"LBR","Liberia"),
    ( 94, 434,"LBY","Libya"),
    ( 95, 662,"LCA","Saint lucia"),
    ( 96, 144,"LKA","Sri lanka"),
    ( 97, 440,"LTU","Lithuania"),
    ( 98, 428,"LVA","Latvia"),
    ( 99, 446,"MAC","Macau"),
    (100, 504,"MAR","Morocco"),
    (101, 498,"MDA","Moldova"),
    (102, 450,"MDG","Madagascar"),
    (103, 462,"MDV","Maldives"),
    (104, 484,"MEX","Mexico"),
    (105, 584,"MHL","Marshall islands"),
    (106, 807,"MKD","Macedonia"),
    (107, 466,"MLI","Mali"),
    (108, 470,"MLT","Malta"),
    (109, 104,"MMR","Myanmar"),
    (110, 496,"MNG","Mongolia"),
    (111, 508,"MOZ","Mozambique"),
    (112, 478,"MRT","Mauritania"),
    (113, 480,"MUS","Mauritius"),
    (114, 454,"MWI","Malawi"),
    (115, 458,"MYS","Malaysia"),
    (116, 540,"NCL","New caledonia"),
    (117, 562,"NER","Niger"),
    (118, 566,"NGA","Nigeria"),
    (119, 558,"NIC","Nicaragua"),
    (120, 528,"NLD","Netherlands"),
    (121, 579,"NOR","Norway"),
    (122, 524,"NPL","Nepal"),
    (123, 554,"NZL","New zealand"),
    (124, 512,"OMN","Oman"),
    (125, 586,"PAK","Pakistan"),
    (126, 591,"PAN","Panama"),
    (127, 604,"PER","Peru"),
    (128, 608,"PHL","Philippines"),
    (129, 598,"PNG","Papua new guinea"),
    (130, 616,"POL","Poland"),
    (131, 408,"PRK","North Korea"),
    (132, 620,"PRT","Portugal"),
    (133, 600,"PRY","Paraguay"),
    (134, 634,"QAT","Qatar"),
    (135, 642,"ROM","Romania"),
    (136, 643,"RUS","Russian federation"),
    (137, 646,"RWA","Rwanda"),
    (138, 682,"SAU","Saudi arabia"),
    (139, 736,"SDN","Sudan"),
    (140, 686,"SEN","Senegal"),
    (141, 702,"SGP","Singapore"),
    (142,  90,"SLB","Solomon islands"),
    (143, 694,"SLE","Sierra leone"),
    (144, 222,"SLV","El salvador"),
    (145, 706,"SOM","Somalia"),
    (146, 740,"SUR","Suriname"),
    (147, 703,"SVK","Slovak republic"),
    (148, 705,"SVN","Slovenia"),
    (149, 752,"SWE","Sweden"),
    (150, 690,"SYC","Seychelles"),
    (151, 760,"SYR","Syria"),
    (152, 148,"TCD","Chad"),
    (153, 768,"TGO","Togo"),
    (154, 764,"THA","Thailand"),
    (155, 762,"TJK","Tajikistan"),
    (156, 795,"TKM","Turkmenistan"),
    (157, 780,"TTO","Trinidad and tobago"),
    (158, 788,"TUN","Tunisia"),
    (159, 792,"TUR","Turkey"),
    (160, 490,"TWN","Taiwan"),
    (161, 834,"TZA","Tanzania"),
    (162, 800,"UGA","Uganda"),
    (163, 804,"UKR","Ukraine"),
    (164, 858,"URY","Uruguay"),
    (165, 842,"USA","United States"),
    (166, 860,"UZB","Uzbekistan"),
    (167, 670,"VCT","Saint vincent and the grenadines"),
    (168, 862,"VEN","Venezuela"),
    (169,  92,"VGB","Virgin Islands"),
    (170, 704,"VNM","Viet nam"),
    (171, 548,"VUT","Vanuatu"),
    (172, 882,"WSM","Samoa"),
    (173, 887,"YEM","Yemen"),
    (174, 891,"YUG","Serbia and Montenegro"),
    (175, 711,"ZAF","South Africa"),
    (176, 180,"ZAR","Congo Dem. Rep."),
    (177, 894,"ZMB","Zambia"),
    (178, 716,"ZWE","Zimbabwe"),
]"""
COUNTRY_PANEL = [
    (  1, 533,"ABW","Aruba"),               (  2,   4,"AFG","Afghanistan"),
    (  3,  24,"AGO","Angola"),              (  4,   8,"ALB","Albania"),
   (  6, 784,"ARE","United Arab Emirates"),
    (  7,  32,"ARG","Argentina"),           (  8,  51,"ARM","Armenia"),
    (  9,  28,"ATG","Antigua and Barbuda"), ( 10,  36,"AUS","Australia"),
    ( 11,  40,"AUT","Austria"),             ( 12,  31,"AZE","Azerbaijan"),
    ( 13, 108,"BDI","Burundi"),             ( 14, 204,"BEN","Benin"),
    ( 15, 854,"BFA","Burkina Faso"),        ( 16,  50,"BGD","Bangladesh"),
    ( 17, 100,"BGR","Bulgaria"),            ( 18,  48,"BHR","Bahrain"),
    ( 19,  70,"BIH","Bosnia and Herzegovina"),( 20, 112,"BLR","Belarus"),
     ( 22,  84,"BLZ","Belize"),
    ( 23,  60,"BMU","Bermuda"),             ( 24,  68,"BOL","Bolivia"),
    ( 25,  76,"BRA","Brazil"),             ( 26,  52,"BRB","Barbados"),
    ( 27,  96,"BRN","Brunei Darussalam"),  ( 28,  64,"BTN","Bhutan"),
    ( 29, 140,"CAF","Central African Republic"),( 30, 124,"CAN","Canada"),
    ( 31, 757,"CHE","Switzerland"),        ( 32, 152,"CHL","Chile"),
    ( 33, 156,"CHN","China"),              ( 34, 384,"CIV","Cote d'Ivoire"),
    ( 35, 120,"CMR","Cameroon"),           ( 36, 178,"COG","Congo"),
    ( 37, 170,"COL","Colombia"),           ( 38, 188,"CRI","Costa Rica"),
    ( 39, 192,"CUB","Cuba"),               ( 40, 136,"CYM","Cayman Islands"),
    ( 41, 196,"CYP","Cyprus"),             ( 42, 203,"CZE","Czech Republic"),
    ( 43, 276,"DEU","Germany"),            ( 44, 212,"DMA","Dominica"),
    ( 45, 208,"DNK","Denmark"),            ( 46, 214,"DOM","Dominican Republic"),
    ( 47,  12,"DZA","Algeria"),            ( 48, 218,"ECU","Ecuador"),
    ( 49, 818,"EGY","Egypt"),              ( 50, 724,"ESP","Spain"),
    ( 51, 233,"EST","Estonia"),            ( 52, 231,"ETH","Ethiopia"),
    ( 53, 246,"FIN","Finland"),            ( 54, 242,"FJI","Fiji"),
    ( 55, 251,"FRA","France"),             ( 56, 583,"FSM","Micronesia"),
    ( 57, 266,"GAB","Gabon"),              ( 58, 826,"GBR","United Kingdom"),
    ( 59, 268,"GEO","Georgia"),            ( 60, 288,"GHA","Ghana"),
    ( 61, 324,"GIN","Guinea"),             ( 62, 270,"GMB","Gambia"),
    ( 63, 624,"GNB","Guinea-Bissau"),      ( 64, 226,"GNQ","Equatorial Guinea"),
    ( 65, 300,"GRC","Greece"),             ( 66, 320,"GTM","Guatemala"),
    ( 67, 328,"GUY","Guyana"),             ( 68, 344,"HKG","Hong Kong"),
    ( 69, 340,"HND","Honduras"),           ( 70, 191,"HRV","Croatia"),
    ( 71, 332,"HTI","Haiti"),              ( 72, 348,"HUN","Hungary"),
    ( 73, 360,"IDN","Indonesia"),          ( 74, 699,"IND","India"),
    ( 75, 372,"IRL","Ireland"),            ( 76, 364,"IRN","Iran"),
    ( 77, 368,"IRQ","Iraq"),               ( 78, 352,"ISL","Iceland"),
    ( 79, 376,"ISR","Israel"),             ( 80, 380,"ITA","Italy"),
    ( 81, 388,"JAM","Jamaica"),            ( 82, 400,"JOR","Jordan"),
    ( 83, 392,"JPN","Japan"),              ( 84, 398,"KAZ","Kazakhstan"),
    ( 85, 404,"KEN","Kenya"),              ( 86, 417,"KGZ","Kyrgyzstan"),
    ( 87, 116,"KHM","Cambodia"),           ( 88, 659,"KNA","Saint Kitts and Nevis"),
    ( 89, 410,"KOR","South Korea"),        ( 90, 414,"KWT","Kuwait"),
    ( 91, 418,"LAO","Laos"),               ( 92, 422,"LBN","Lebanon"),
    ( 93, 430,"LBR","Liberia"),            ( 94, 434,"LBY","Libya"),
    ( 95, 662,"LCA","Saint Lucia"),        ( 96, 144,"LKA","Sri Lanka"),
    ( 97, 440,"LTU","Lithuania"),          ( 98, 428,"LVA","Latvia"),
    ( 99, 446,"MAC","Macau"),              (100, 504,"MAR","Morocco"),
    (101, 498,"MDA","Moldova"),            (102, 450,"MDG","Madagascar"),
    (103, 462,"MDV","Maldives"),           (104, 484,"MEX","Mexico"),
    (105, 584,"MHL","Marshall Islands"),   (106, 807,"MKD","Macedonia"),
    (107, 466,"MLI","Mali"),               (108, 470,"MLT","Malta"),
    (109, 104,"MMR","Myanmar"),            (110, 496,"MNG","Mongolia"),
    (111, 508,"MOZ","Mozambique"),         (112, 478,"MRT","Mauritania"),
    (113, 480,"MUS","Mauritius"),          (114, 454,"MWI","Malawi"),
    (115, 458,"MYS","Malaysia"),           (116, 540,"NCL","New Caledonia"),
    (117, 562,"NER","Niger"),              (118, 566,"NGA","Nigeria"),
    (119, 558,"NIC","Nicaragua"),          (120, 528,"NLD","Netherlands"),
    (121, 579,"NOR","Norway"),             (122, 524,"NPL","Nepal"),
    (123, 554,"NZL","New Zealand"),        (124, 512,"OMN","Oman"),
    (125, 586,"PAK","Pakistan"),           (126, 591,"PAN","Panama"),
    (127, 604,"PER","Peru"),               (128, 608,"PHL","Philippines"),
    (129, 598,"PNG","Papua New Guinea"),   (130, 616,"POL","Poland"),
    (131, 408,"PRK","North Korea"),        (132, 620,"PRT","Portugal"),
    (133, 600,"PRY","Paraguay"),           (134, 634,"QAT","Qatar"),
    (135, 642,"ROM","Romania"),            (136, 643,"RUS","Russian Federation"),
    (137, 646,"RWA","Rwanda"),             (138, 682,"SAU","Saudi Arabia"),
    (140, 686,"SEN","Senegal"),
    (141, 702,"SGP","Singapore"),          (142,  90,"SLB","Solomon Islands"),
    (143, 694,"SLE","Sierra Leone"),       (144, 222,"SLV","El Salvador"),
    (145, 706,"SOM","Somalia"),            (146, 740,"SUR","Suriname"),
    (147, 703,"SVK","Slovak Republic"),    (148, 705,"SVN","Slovenia"),
    (149, 752,"SWE","Sweden"),             (150, 690,"SYC","Seychelles"),
    (151, 760,"SYR","Syria"),              (152, 148,"TCD","Chad"),
    (153, 768,"TGO","Togo"),               (154, 764,"THA","Thailand"),
    (155, 762,"TJK","Tajikistan"),         (156, 795,"TKM","Turkmenistan"),
    (157, 780,"TTO","Trinidad and Tobago"),(158, 788,"TUN","Tunisia"),
    (159, 792,"TUR","Turkey"),             (160, 490,"TWN","Taiwan"),
    (161, 834,"TZA","Tanzania"),           (162, 800,"UGA","Uganda"),
    (163, 804,"UKR","Ukraine"),            (164, 858,"URY","Uruguay"),
    (165, 842,"USA","United States"),      (166, 860,"UZB","Uzbekistan"),
    (167, 670,"VCT","Saint Vincent and the Grenadines"),
    (168, 862,"VEN","Venezuela"),          (169,  92,"VGB","Virgin Islands"),
    (170, 704,"VNM","Viet Nam"),           (171, 548,"VUT","Vanuatu"),
    (172, 882,"WSM","Samoa"),              (173, 887,"YEM","Yemen"),

    (176, 180,"ZAR","Congo Dem. Rep."),
    (177, 894,"ZMB","Zambia"),             (178, 716,"ZWE","Zimbabwe"),
    (179, 56, "BEL","Belgium"),            (180, 442, "LUX","Luxembourg")
]

# Belgium(56) + Luxembourg(442) → BLX
NUMERIC_TO_ISO3 = {row[1]: row[2] for row in COUNTRY_PANEL}

ISO3_TO_ID  = {row[2]: row[0] for row in COUNTRY_PANEL}
PANEL_ISO3  = [row[2] for row in COUNTRY_PANEL]

N_PANEL = 175
N1      = 174   # N_PANEL - 1

COL_ORDER = [
    "t","id","i","iso3","country",
    "Out-degree","In-degree",
    "Out-strength","In-strength",
    "Out-degree_percent","In-degree percent",
    "Out-strength-percent","In-strength_percent",
    "Out-closenness","In-closenness",
    "W-Out-closenness","W-In-closenness",
    "Out-eigenvector","In-eigenvector",
    "W-Out-eigenvector","W-In-eigenvector",
]

# ══════════════════════════════════════════════════════════════
# FUNCTION 1 — Load & filter one year of BACI HS17 data
# ══════════════════════════════════════════════════════════════
def load_baci_year(year):
    pattern = os.path.join(BACI_DIR, f"BACI_HS17_Y{year}_V*.csv")
    files   = glob.glob(pattern)
    if not files:
        raise FileNotFoundError(
            f"No BACI HS17 file for {year}.\n"
            f"Expected: {pattern}\n"
            f"Download from: https://www.cepii.fr/DATA_DOWNLOAD/baci/data/BACI_HS17_V202601.zip\n"
            f"Unzip into: {BACI_DIR}"
        )

    print(f"  Loading {os.path.basename(files[0])} ...", end=" ", flush=True)
    df = pd.read_csv(files[0], dtype={"k": str, "i": int, "j": int, "v": float})
    total_rows = len(df)

    # ── KEY STEP: keep only semiconductor product codes ──────────────
    # Pad k to 6 digits (BACI sometimes stores as integer dropping leading zeros)
    df["k"] = df["k"].str.zfill(6)
    df = df[df["k"].isin(SEMICON_CODES)]
    print(f"{total_rows:,} total rows → {len(df):,} semiconductor rows "
          f"({len(df)/total_rows*100:.1f}% of year)")

    if len(df) == 0:
        print(f"  WARNING: No semiconductor trade found for {year}. "
              f"Check that product codes match HS17 revision.")
        return pd.DataFrame(columns=["exp","imp","val"])

    # Map numeric codes → ISO3
    df["exp"] = df["i"].map(NUMERIC_TO_ISO3)
    df["imp"] = df["j"].map(NUMERIC_TO_ISO3)
    df = df.dropna(subset=["exp","imp"])
    df = df[df["exp"] != df["imp"]]   # no self-loops

    # Aggregate all 59 semiconductor products → one bilateral total
    bilateral = (df.groupby(["exp","imp"])["v"]
                   .sum().reset_index()
                   .rename(columns={"v":"val"}))
    print(f"  → {len(bilateral):,} bilateral flows | "
          f"{bilateral['exp'].nunique()} exporters | "
          f"{bilateral['imp'].nunique()} importers")
    return bilateral


# ══════════════════════════════════════════════════════════════
# FUNCTION 2 — Build directed weighted graph
# ══════════════════════════════════════════════════════════════
def build_graph(bilateral):
    G = nx.DiGraph()
    G.add_nodes_from(PANEL_ISO3)
    for _, row in bilateral.iterrows():
        if row["exp"] in ISO3_TO_ID and row["imp"] in ISO3_TO_ID:
            G.add_edge(row["exp"], row["imp"], weight=float(row["val"]))
    return G


# ══════════════════════════════════════════════════════════════
# FUNCTION 3 — Degree + Strength
# ══════════════════════════════════════════════════════════════
def compute_degree_strength(G, active):
    M           = G.number_of_edges()
    total_trade = sum(d["weight"] for _,_,d in G.edges(data=True))
    results     = {}
    for n in active:
        ol  = G.out_degree(n)
        il  = G.in_degree(n)
        os_ = sum(d["weight"] for _,_,d in G.out_edges(n, data=True))
        is_ = sum(d["weight"] for _,_,d in G.in_edges(n,  data=True))
        results[n] = {
            "Out-degree":           ol  / N1,
            "In-degree":            il  / N1,
            "Out-strength":         os_ / N1,
            "In-strength":          is_ / N1,
            "Out-degree_percent":   ol  / M           * 100 if M > 0 else 0,
            "In-degree percent":    il  / M           * 100 if M > 0 else 0,
            "Out-strength-percent": os_ / total_trade * 100 if total_trade > 0 else 0,
            "In-strength_percent":  is_ / total_trade * 100 if total_trade > 0 else 0,
        }
    return results


# ══════════════════════════════════════════════════════════════
# FUNCTION 4 — Unweighted closeness
# ══════════════════════════════════════════════════════════════
def compute_closeness_unweighted(G, active):
    G_rev = G.reverse(copy=False)
    results = {}
    for n in active:
        sp_out = nx.single_source_shortest_path_length(G, n)
        d_out  = sum(sp_out.get(j, N_PANEL) for j in active if j != n)
        sp_in  = nx.single_source_shortest_path_length(G_rev, n)
        d_in   = sum(sp_in.get(j,  N_PANEL) for j in active if j != n)
        results[n] = {
            "Out-closenness": N1 / d_out if d_out > 0 else 0.0,
            "In-closenness":  N1 / d_in  if d_in  > 0 else 0.0,
        }
    return results


# ══════════════════════════════════════════════════════════════
# FUNCTION 5 — Weighted closeness (Newman/Opsahl)
# omega_ij = N * W_ij / ΣW   →   dist = 1/omega
# ══════════════════════════════════════════════════════════════
def compute_closeness_weighted(G, active):
    total_W = sum(d["weight"] for _,_,d in G.edges(data=True))
    if total_W == 0:
        return {n: {"W-Out-closenness": 0.0, "W-In-closenness": 0.0} for n in active}

    Gd = nx.DiGraph()
    Gd.add_nodes_from(active)
    for u, v, data in G.edges(data=True):
        omega = N_PANEL * data["weight"] / total_W
        Gd.add_edge(u, v, weight=1.0 / omega if omega > 0 else 1e12)
    Gd_rev = Gd.reverse(copy=False)

    results = {}
    for n in active:
        try:
            sp    = dict(nx.single_source_dijkstra_path_length(Gd, n, weight="weight"))
            d_sum = sum(sp.get(j, 1e12) for j in active if j != n)
            wc_out = N1 / d_sum if d_sum > 0 else 0.0
        except Exception:
            wc_out = 0.0
        try:
            sp    = dict(nx.single_source_dijkstra_path_length(Gd_rev, n, weight="weight"))
            d_sum = sum(sp.get(j, 1e12) for j in active if j != n)
            wc_in  = N1 / d_sum if d_sum > 0 else 0.0
        except Exception:
            wc_in  = 0.0
        results[n] = {"W-Out-closenness": wc_out, "W-In-closenness": wc_in}
    return results


# ══════════════════════════════════════════════════════════════
# FUNCTION 6 — Eigenvector centrality (L2-normalised)
# ══════════════════════════════════════════════════════════════
def _leading_eigvec_l2(mat):
    n = mat.shape[0]
    try:
        _, vecs = eigs(mat.astype(float), k=1, which="LM",
                       maxiter=10_000, tol=1e-10)
        v    = np.abs(vecs[:, 0].real)
        norm = np.linalg.norm(v)
        return v / norm if norm > 1e-12 else np.zeros(n)
    except Exception:
        return np.zeros(n)


def compute_eigenvector(G, active, weighted=False):
    n   = len(active)
    idx = {nd: k for k, nd in enumerate(active)}
    rows, cols, vals = [], [], []
    for u, v, data in G.edges(data=True):
        if u in idx and v in idx:
            rows.append(idx[u])
            cols.append(idx[v])
            vals.append(data["weight"] if weighted else 1.0)
    if not vals:
        zeros = {nd: 0.0 for nd in active}
        return zeros, zeros
    A = csr_matrix((vals, (rows, cols)), shape=(n, n), dtype=float)
    ev_out = {nd: float(_leading_eigvec_l2(A)[idx[nd]])   for nd in active}
    ev_in  = {nd: float(_leading_eigvec_l2(A.T)[idx[nd]]) for nd in active}
    return ev_out, ev_in


# ══════════════════════════════════════════════════════════════
# FUNCTION 7 — Process one complete year
# ══════════════════════════════════════════════════════════════
def process_year(year):
    bilateral = load_baci_year(year)

    if len(bilateral) == 0:
        # Return zeros for all countries if no semiconductor trade found
        records = []
        for cid, cnum, iso3, cname in COUNTRY_PANEL:
            row = {"t": year, "id": cid, "i": cnum, "iso3": iso3, "country": cname}
            for col in COL_ORDER[5:]:
                row[col] = 0.0
            records.append(row)
        return pd.DataFrame(records)[COL_ORDER]

    G = build_graph(bilateral)
    active = [iso for iso in PANEL_ISO3
              if G.out_degree(iso) > 0 or G.in_degree(iso) > 0]
    M = G.number_of_edges()
    print(f"  Active: {len(active)}/{N_PANEL} countries | Edges: {M:,}")

    print("  [1/4] Degree + Strength ...",          end=" ", flush=True)
    deg  = compute_degree_strength(G, active);       print("done")

    print("  [2/4] Unweighted closeness ...",        end=" ", flush=True)
    cl_u = compute_closeness_unweighted(G, active);  print("done")

    print("  [3/4] Weighted closeness (slowest) ...", end=" ", flush=True)
    cl_w = compute_closeness_weighted(G, active);    print("done")

    print("  [4/4] Eigenvector centrality ...",      end=" ", flush=True)
    ev_u_out, ev_u_in = compute_eigenvector(G, active, weighted=False)
    ev_w_out, ev_w_in = compute_eigenvector(G, active, weighted=True)
    print("done")

    records = []
    for cid, cnum, iso3, cname in COUNTRY_PANEL:
        row = {"t": year, "id": cid, "i": cnum, "iso3": iso3, "country": cname}
        if iso3 in active:
            row.update(deg[iso3])
            row.update(cl_u[iso3])
            row.update(cl_w[iso3])
            row["Out-eigenvector"]   = ev_u_out[iso3]
            row["In-eigenvector"]    = ev_u_in[iso3]
            row["W-Out-eigenvector"] = ev_w_out[iso3]
            row["W-In-eigenvector"]  = ev_w_in[iso3]
        else:
            for col in COL_ORDER[5:]:
                row[col] = 0.0
        records.append(row)

    return pd.DataFrame(records)[COL_ORDER]


# ══════════════════════════════════════════════════════════════
# MAIN LOOP
# ══════════════════════════════════════════════════════════════
print("=" * 60)
print(f"  Semiconductor Trade Network Centrality")
print(f"  HS17 | {len(SEMICON_CODES)} products | {YEARS[0]}–{YEARS[-1]}")
print(f"  Countries : {N_PANEL} (fixed panel)")
print(f"  Output    : {OUTPUT_CSV}")
print("=" * 60)

all_dfs = []
skipped = []
t_start = time.time()

for year in YEARS:
    print(f"\n{'─'*60}")
    print(f"  YEAR {year}")
    print(f"{'─'*60}")
    t_year = time.time()
    try:
        df_year = process_year(year)
        all_dfs.append(df_year)

        # Save per-year backup immediately
        backup = f"/content/drive/MyDrive/semiconductor_centrality_{year}.csv"
        df_year.to_csv(backup, sep=";", index=False, float_format="%.7g")

        elapsed   = time.time() - t_year
        done      = YEARS.index(year) + 1
        remaining = len(YEARS) - done
        eta       = (time.time() - t_start) / done * remaining / 60
        print(f"  ✓  Saved → {backup}")
        print(f"     Time: {elapsed/60:.1f} min | ETA: {eta:.0f} min remaining")

    except FileNotFoundError as e:
        print(f"  ✗  SKIPPED — {e}")
        skipped.append(year)
    except Exception as e:
        print(f"  ✗  ERROR — {e}")
        import traceback; traceback.print_exc()
        skipped.append(year)

# ══════════════════════════════════════════════════════════════
# SAVE FINAL CSV
# ══════════════════════════════════════════════════════════════
if all_dfs:
    final = pd.concat(all_dfs, ignore_index=True).sort_values(["t","id"])
    final.to_csv(OUTPUT_CSV, sep=";", index=False, float_format="%.7g")
    total_min = (time.time() - t_start) / 60
    print(f"\n{'='*60}")
    print(f"  COMPLETE!")
    print(f"  File      : {OUTPUT_CSV}")
    print(f"  Rows      : {len(final):,}  ({len(all_dfs)} yrs × {N_PANEL} countries)")
    print(f"  Columns   : {len(final.columns)}")
    print(f"  Products  : {len(SEMICON_CODES)} semiconductor HS17 codes")
    print(f"  Time      : {total_min:.1f} minutes total")
    if skipped:
        print(f"  Skipped   : {skipped}")
    print("=" * 60)
else:
    print("\nERROR: Nothing processed. Check BACI_DIR path.")


# ══════════════════════════════════════════════════════════════
# OPTIONAL: Per-product breakdown (run separately if needed)
# Shows which products dominate semiconductor trade each year
# ══════════════════════════════════════════════════════════════
# Uncomment and run in a separate cell after main loop finishes:
#
# BACI_DIR_CHECK = "/content/baci_hs17"
# year_check = 2022
# pattern = os.path.join(BACI_DIR_CHECK, f"BACI_HS17_Y{year_check}_V*.csv")
# df_raw = pd.read_csv(glob.glob(pattern)[0], dtype={"k":str,"v":float})
# df_raw["k"] = df_raw["k"].str.zfill(6)
# df_semi = df_raw[df_raw["k"].isin(SEMICON_CODES)]
# by_product = (df_semi.groupby("k")["v"].sum()
#                      .sort_values(ascending=False)
#                      .reset_index())
# by_product["share_pct"] = by_product["v"] / by_product["v"].sum() * 100
# print(f"\nTop semiconductor products by trade value ({year_check}):")
# print(by_product.head(20).to_string(index=False))
```
